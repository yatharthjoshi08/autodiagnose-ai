```python
# Function to calculate mean:
def calculate_mean(data):
    total = 0 # Sum of all values
    count = 0 # Number of values
    for value in data:
        # Skip missing values or non-numerical values:
        if pd.isna(value):
            continue
        total += value
        count += 1
    mean = total/count
    return mean
```

```python
# Calculating mean for all numerical columns:
for column in numerical_columns:
    data = df[column].tolist() # Get data from the current column
    mean = calculate_mean(data)
    print("Mean of", column, "=", mean)
```

```python
# Function to calculate median:
def calculate_median(data):
    values = data.copy() # Creating a copy of data
    valid_values = []
    # This part is to make a valid list which does not have null values so as to compute median
    for value in values:
        # Skip the missing value
        if pd.isna(value):
            continue
        valid_values.append(value)
    # Using only valid values:
    values = valid_values
    values.sort() # Sorting the values
    n = len(values)
    # If number of values is odd:
    if n%2 != 0:
        median = values[n//2] # here n//2 since index start from 0
    # If number of values is even:
    else:
        middle1 = values[(n//2)-1]
        middle2 = values[n//2]
        median = (middle1 + middle2)/2

    return median
```

```python
# Calculate the median for each numerical column
for column in numerical_columns:
    data = df[column].tolist() # Get the data from the current column
    median = calculate_median(data)
    print("Median of", column, "=", median)
```

```python
# Function to calculate mode:
def calculate_mode(data):
    # Empty dictionary to store frequency of each value:
    freq = {}
    for value in data:
        # Check if null value if yes skip:
        if pd.isna(value):
            continue
        # Check if value already present:
        if value in freq:
            freq[value] += 1 # Already present so freq inc by 1
        else:
            freq[value] = 1 # Appear for first time so freq = 1
    # Store the heighest freq:
    highest_freq = 0
    for value in freq:
        if freq[value] > highest_freq:
            highest_freq = freq[value]
    # List to store the mode values:
    mode = []
    for value in freq:
        if freq[value] == highest_freq:
            mode.append(value)

    return mode
        
```

```python
# Calculate the mode for each numerical column
for column in numerical_columns:
    data = df[column].tolist()
    mode = calculate_mode(data)
    print("Mode of", column, "=", mode)
```

```python
# Function to calculate Standard Deviation:
def calculate_std_dev(data):
    # Find mean first:
    mean = calculate_mean(data)
    total = 0 # sum of squared difference:
    count = 0 # number of valid values
    for value in data:
        # Skip missing values:
        if pd.isna(value):
            continue
        difference = value - mean
        squared_diff = difference**2
        total += squared_diff
        count += 1
    # Calculate variance:
    variance = total/count
    # Calculate Std Dev:
    stdev = math.sqrt(variance)

    return stdev
```

```python
# Calculate the standard deviation for each numerical column
for column in numerical_columns:
    data = df[column].tolist()
    standard_deviation = calculate_std_dev(data)
    print("Standard Deviation of", column, "=", standard_deviation)
```

```python
rows_before = len(df)

# drop_duplicates() keeps the first copy and removes the repeated ones
# reset_index(drop=True) numbers the rows 0,1,2,... again
df = df.drop_duplicates().reset_index(drop=True)

print("Rows before:", rows_before)
print("Rows after :", len(df))
print("Removed    :", rows_before - len(df))

# Note: Proof. If the repeats were exact copies, every Record_ID is now unique.
print("Record_ID unique now? ->", df['Record_ID'].is_unique)
```

```python
# Text columns = all columns that are not numbers
text_cols = [c for c in df.columns if not pd.api.types.is_numeric_dtype(df[c])]

# Copy of the data before this step (only to count how many values change)
before = df.copy()

for c in text_cols:
    # 1) remove spaces at the start/end and make double spaces single
    clean = df[c].str.strip().str.replace(r'\s+', ' ', regex=True)
    # 2) lower case version is used only to find which values are "the same"
    lower = clean.str.lower()
    # 3) for every lower case value, find its most common original spelling
    best_spelling = {}
    for value in lower.dropna().unique():
        best_spelling[value] = clean[lower == value].value_counts().index[0]
    # 4) write the best spelling back
    df[c] = lower.map(best_spelling)

# Show how many categories each column had before and after (only columns that changed)
rows = []
for c in text_cols:
    rows.append({'Column': c,
                 'values changed': int(((df[c] != before[c]) & df[c].notna()).sum()),
                 'categories before': before[c].nunique(),
                 'categories after': df[c].nunique()})
summary = pd.DataFrame(rows)
summary[summary['values changed'] > 0].sort_values('values changed', ascending=False)
```

```python
# How many '-' cells are there in each column? (only columns that have some)
dash_count = (df_raw == '-').sum()
print("'-' per column:")
print(dash_count[dash_count > 0])

# Replace '-' by NaN everywhere
df = df.replace('-', np.nan)

# 'unknown' and 'not available' (any upper/lower case) -> 'Unknown'
for c in text_cols:
    is_unknown = df[c].str.lower().isin(['unknown', 'not available'])
    df.loc[is_unknown, c] = 'Unknown'
```

```python
# A value like '1.3 hrs' or '250' -> number part first, optional unit letters after it
num_pattern = r'^\s*(-?\d+(?:\.\d+)?)\s*[A-Za-z]*\s*$'
# A value like '04-05-26' or '08-Oct-24' -> three parts joined by - / or .
date_pattern = r'^\s*\d{1,4}[-/.](?:\d{1,2}|[A-Za-z]{3})[-/.]\d{2,4}\s*$'

dtypes_before = df.dtypes.astype(str)

for c in text_cols:
    values = df[c].dropna()
    share_num = values.str.match(num_pattern).mean()      # share of values that look like numbers
    share_date = values.str.match(date_pattern).mean()    # share of values that look like dates
    if share_num >= 0.90:
        df[c] = pd.to_numeric(df[c].str.extract(num_pattern)[0])
        print(c, "-> number")
    elif share_date >= 0.90:
        # Show the formats found: every digit becomes 9, month names become Mon
        shape = values.str.replace(r'\d', '9', regex=True).str.replace(r'[A-Za-z]{3}', 'Mon', regex=True)
        print(c, "-> date. Formats found:")
        print(shape.value_counts().to_string())
```

```python
# Which part is the day and which is the month? A part bigger than 12 cannot be a month.
dates = df['Inspection_Date'].dropna().astype(str).str.strip()
shape = dates.str.replace(r'\d', '9', regex=True)
for f in ['99-99-99', '99.99.9999', '99/99/9999']:
    sep = f[2]                                            # the separator is the 3rd character
    parts = dates[shape == f].str.split(sep, regex=False, expand=True).astype(int)
    print(f, "-> first part > 12:", (parts[0] > 12).sum(), "| second part > 12:", (parts[1] > 12).sum())
# Result: the first part goes above 12 in '99-99-99' and '99.99.9999' (so it is the DAY),
# the second part goes above 12 in '99/99/9999' (so it is the DAY there, first part is the MONTH).

# So the 4 formats are:
date_formats = ['%d-%m-%y',    # 04-05-26
                '%d.%m.%Y',    # 29.12.2025
                '%m/%d/%Y',    # 07/29/2025
                '%d-%b-%y']    # 08-Oct-24

def parse_date(text):
    # Try every format, the first one that works wins
    for f in date_formats:
        try:
            return pd.to_datetime(text, format=f)
        except (ValueError, TypeError):
            pass
    return pd.NaT      # no format worked -> missing date

df['Inspection_Date'] = df['Inspection_Date'].astype(str).str.strip().map(parse_date)

# Note: Proof. The date column must have no new missing values and a sensible date range.
print("Missing dates:", df['Inspection_Date'].isna().sum())
print("Date range   :", df['Inspection_Date'].min().date(), "to", df['Inspection_Date'].max().date())

# Compare data types before and after
dtypes_after = df.dtypes.astype(str)
change = pd.DataFrame({'before': dtypes_before, 'after': dtypes_after})
change[change['before'] != change['after']]
```

```python
# Possible range of each numeric column in real life (lowest, highest)
valid_range = {
    'Year'                       : (1995, 2026),
    'Mileage_KM'                 : (1, 400000),
    'Engine_Displacement_CC'     : (600, 6000),
    'Fuel_Efficiency_kmpl'       : (1, 60),
    'Estimated_Repair_Time_Hours': (0.1, 40),
    'Customer_Rating'            : (1, 5),
}

rows = []
for c in valid_range:
    low, high = valid_range[c]
    x = df[c].dropna()                          # only the values that exist

    # Method 1: IQR rule
    q1 = x.quantile(0.25)
    q3 = x.quantile(0.75)
    iqr = q3 - q1
    by_iqr = (x < q1 - 1.5 * iqr) | (x > q3 + 1.5 * iqr)

    # Method 2: Z-score rule
    z = (x - x.mean()) / x.std()
    by_z = z.abs() > 3

    # Method 3: domain rule (this is the truth we compare with)
    error = (x < low) | (x > high)

    rows.append({'Column': c,
                 'IQR flagged': by_iqr.sum(),
                 'IQR flagged genuine': (by_iqr & ~error).sum(),
                 'Z flagged': by_z.sum(),
                 'Z flagged genuine': (by_z & ~error).sum(),
                 'Real errors (domain)': error.sum(),
                 'Errors missed by IQR': (error & ~by_iqr).sum(),
                 'Errors missed by Z': (error & ~by_z).sum()})
pd.DataFrame(rows)
```

```python
# Rows where every column used in the test is known
known = df.dropna(subset=['Year', 'Mileage_KM', 'Engine_Displacement_CC', 'Fuel_Efficiency_kmpl',
                          'Estimated_Repair_Time_Hours', 'Model', 'Fuel', 'Diagnosis']).reset_index(drop=True)

def split_data(data, seed):
    # 10% of rows are hidden (test), the other 90% are used to learn the fill value (train)
    test = data.sample(frac=0.1, random_state=seed)
    train = data.drop(test.index)
    return train, test

def mae_fill_global(train, test, col, how):
    # one single value for all rows: mean or median of the train rows
    value = train[col].mean() if how == 'mean' else train[col].median()
    return abs(test[col] - value).mean()

def mae_fill_group(train, test, col, keys):
    # median of the train rows of the same group (e.g. same Model + Fuel)
    group_median = train.groupby(keys)[col].median().rename('guess').reset_index()
    guess = test[keys].merge(group_median, on=keys, how='left')['guess']
    guess = guess.fillna(train[col].median()).to_numpy()      # group not seen -> overall median
    return abs(test[col].to_numpy() - guess).mean()

knn_cols = ['Year', 'Mileage_KM', 'Engine_Displacement_CC', 'Fuel_Efficiency_kmpl']

def mae_fill_knn(train, test, col):
    both = pd.concat([train[knn_cols], test[knn_cols]])
    mean, std = both.mean(), both.std()
    scaled = (both - mean) / std                  # KNN needs all columns on the same scale
    scaled.loc[test.index, col] = np.nan          # hide the test values
    filled = KNNImputer(n_neighbors=5).fit_transform(scaled)
    filled = pd.DataFrame(filled, index=scaled.index, columns=knn_cols)
    guess = filled.loc[test.index, col] * std[col] + mean[col]       # back to the real scale
    return abs(test[col].to_numpy() - guess.to_numpy()).mean()
```

```python
# Every row: (column, method name, function to call)
E, F, R_ = 'Engine_Displacement_CC', 'Fuel_Efficiency_kmpl', 'Estimated_Repair_Time_Hours'
tests = [
    (E, 'global median',          lambda tr, te: mae_fill_global(tr, te, E, 'median')),
    (E, 'median by Model',        lambda tr, te: mae_fill_group(tr, te, E, ['Model'])),
    (E, 'median by Model+Fuel',   lambda tr, te: mae_fill_group(tr, te, E, ['Model', 'Fuel'])),
    (F, 'global median',          lambda tr, te: mae_fill_global(tr, te, F, 'median')),
    (F, 'median by Fuel',         lambda tr, te: mae_fill_group(tr, te, F, ['Fuel'])),
    (F, 'median by Model+Fuel',   lambda tr, te: mae_fill_group(tr, te, F, ['Model', 'Fuel'])),
    ('Mileage_KM', 'mean',        lambda tr, te: mae_fill_global(tr, te, 'Mileage_KM', 'mean')),
    ('Mileage_KM', 'median',      lambda tr, te: mae_fill_global(tr, te, 'Mileage_KM', 'median')),
    ('Mileage_KM', 'KNN',         lambda tr, te: mae_fill_knn(tr, te, 'Mileage_KM')),
    ('Year', 'mean',              lambda tr, te: mae_fill_global(tr, te, 'Year', 'mean')),
    ('Year', 'median',            lambda tr, te: mae_fill_global(tr, te, 'Year', 'median')),
    ('Year', 'KNN',               lambda tr, te: mae_fill_knn(tr, te, 'Year')),
    (R_, 'global median',         lambda tr, te: mae_fill_global(tr, te, R_, 'median')),
    (R_, 'median by Diagnosis',   lambda tr, te: mae_fill_group(tr, te, R_, ['Diagnosis'])),
]
results = []
for col, method, function in tests:
    errors = []
    for seed in range(5):
        train, test = split_data(known, seed)
        errors.append(function(train, test))
    results.append({'Column': col, 'Method': method, 'MAE': round(np.mean(errors), 2)})
pd.DataFrame(results)
```

```python
# Same test for text columns: how often is the group-wise mode (most common value) correct?
def accuracy_test(data, col, keys):
    d = data.dropna(subset=[col] + keys).reset_index(drop=True)
    plain, grouped = [], []
    for seed in range(5):
        train, test = split_data(d, seed)
        overall_mode = train[col].mode().iloc[0]
        plain.append((test[col] == overall_mode).mean())
        group_mode = train.groupby(keys)[col].agg(lambda s: s.mode().iloc[0]).rename('guess').reset_index()
        guess = test[keys].merge(group_mode, on=keys, how='left')['guess'].fillna(overall_mode).to_numpy()
        grouped.append((guess == test[col].to_numpy()).mean())
    return round(np.mean(plain) * 100, 1), round(np.mean(grouped) * 100, 1)

known_city = df[df['City'] != 'Unknown']     # 'Unknown' city says nothing about the state
cat_tests = [
    ('Make',               ['Model'],                                          df,         'Make by Model'),
    ('Model',              ['Make', 'Vehicle_Type'],                           df,         'Model by Make+Vehicle_Type'),
    ('Model',              ['Make', 'Vehicle_Type', 'Engine_Displacement_CC'], df,         'Model by Make+Vehicle_Type+CC'),
    ('Registration_State', ['City'],                                           known_city, 'State by City'),
    ('Fuel',               ['Model', 'Engine_Displacement_CC'],                df,         'Fuel by Model+CC'),
    ('Transmission',       ['Make', 'Fuel'],                                   df,         'Transmission by Make+Fuel'),
]
rows = []
for col, keys, data, label in cat_tests:
    plain, grouped = accuracy_test(data, col, keys)
    rows.append({'Column': col, 'Method': label,
                 'Global mode accuracy %': plain, 'Group-wise mode accuracy %': grouped})
pd.DataFrame(rows)
```

```python
# Two small helper functions used many times below
def fill_with_group_mode(table, col, keys, learn_rows=None):
    # learn the most common value of col for every group of keys, then fill the missing cells
    learn = table if learn_rows is None else table[learn_rows]
    best = learn.groupby(keys)[col].agg(lambda s: s.mode().iloc[0] if s.notna().any() else np.nan)
    guess = table[keys].merge(best.rename('guess').reset_index(), on=keys, how='left')['guess']
    return table[col].fillna(pd.Series(guess.to_numpy(), index=table.index))

def fill_with_group_median(table, col, keys):
    # same idea for numbers, using the median
    best = table.groupby(keys)[col].median()
    guess = table[keys].merge(best.rename('guess').reset_index(), on=keys, how='left')['guess']
    return table[col].fillna(pd.Series(guess.to_numpy(), index=table.index))
```

```python
# (a) Structural missing -> constant categories
df['Symptom_2'] = df['Symptom_2'].fillna('No_Symptom')
df['Symptom_3'] = df['Symptom_3'].fillna('No_Symptom')
df['Followup_Question_2'] = df['Followup_Question_2'].fillna('No_Followup')
df['Followup_Answer_2'] = df['Followup_Answer_2'].fillna('No_Followup')
df['OBD_Code'] = df['OBD_Code'].fillna('No_Code')

# (b) Text columns: group-wise mode (the group is chosen by logic and proved in Step 8)
df['Make'] = fill_with_group_mode(df, 'Make', ['Model'])                  # one Model belongs to one Make
df['Registration_State'] = fill_with_group_mode(df, 'Registration_State', ['City'],
                                                learn_rows=df['City'] != 'Unknown')   # one City belongs to one State
df['Model'] = fill_with_group_mode(df, 'Model', ['Make', 'Vehicle_Type', 'Engine_Displacement_CC'])
df['Model'] = fill_with_group_mode(df, 'Model', ['Make', 'Vehicle_Type'])     # rare: CC was also missing
df['Model'] = fill_with_group_mode(df, 'Model', ['Vehicle_Type'])             # rare: Make was also missing
df['Make'] = fill_with_group_mode(df, 'Make', ['Model'])                      # Make again, now Model is known
df['Fuel'] = fill_with_group_mode(df, 'Fuel', ['Model', 'Engine_Displacement_CC'])

# (c) Last fallback for text: most common value, or 'Unknown' where a mode makes no sense
for c in ['Fuel', 'Transmission', 'Owner_Number', 'Registration_State', 'Model', 'Make']:
    df[c] = df[c].fillna(df[c].mode().iloc[0])
for c in ['Insurance_Status', 'Service_History', 'City']:
    df[c] = df[c].fillna('Unknown')
```

```python
# (d) Numbers
# Engine CC and Fuel Efficiency: median of the same Model + Fuel (then wider groups as fallback)
df['Engine_Displacement_CC'] = fill_with_group_median(df, 'Engine_Displacement_CC', ['Model', 'Fuel'])
df['Engine_Displacement_CC'] = fill_with_group_median(df, 'Engine_Displacement_CC', ['Model'])
df['Engine_Displacement_CC'] = df['Engine_Displacement_CC'].fillna(df['Engine_Displacement_CC'].median())
df['Fuel_Efficiency_kmpl'] = fill_with_group_median(df, 'Fuel_Efficiency_kmpl', ['Model', 'Fuel'])
df['Fuel_Efficiency_kmpl'] = fill_with_group_median(df, 'Fuel_Efficiency_kmpl', ['Fuel'])
df['Fuel_Efficiency_kmpl'] = df['Fuel_Efficiency_kmpl'].fillna(df['Fuel_Efficiency_kmpl'].median())

# Year and Mileage: KNN (k = 5) on scaled columns
mean, std = df[knn_cols].mean(), df[knn_cols].std()
scaled = (df[knn_cols] - mean) / std
filled = KNNImputer(n_neighbors=5).fit_transform(scaled)
filled = pd.DataFrame(filled * std.values + mean.values, columns=knn_cols, index=df.index)
df['Year'] = filled['Year'].round()
df['Mileage_KM'] = filled['Mileage_KM'].round()

# Repair time: median of the same Diagnosis. Customer rating: overall median.
df['Estimated_Repair_Time_Hours'] = fill_with_group_median(df, 'Estimated_Repair_Time_Hours', ['Diagnosis'])
df['Estimated_Repair_Time_Hours'] = df['Estimated_Repair_Time_Hours'].fillna(df['Estimated_Repair_Time_Hours'].median())
df['Customer_Rating'] = df['Customer_Rating'].fillna(df['Customer_Rating'].median())

# Whole-number columns get the whole-number type
for c in ['Year', 'Mileage_KM', 'Engine_Displacement_CC', 'Customer_Rating']:
    df[c] = df[c].astype(int)
```

# EXPERIMENT 3:
# PROBABILITY DISTRIBUTIONS AND HYPOTHESIS TESTS

```python
import numpy as np                  # numbers and formulas
import pandas as pd                 # tables
import matplotlib.pyplot as plt     # graphs
from scipy import stats             # only for p-values and one-line cross-checks
import statsmodels.formula.api as smf   # only to cross-check the two-way ANOVA
import statsmodels.api as sm

df = pd.read_csv('vehicle_diagnostic_clean.csv')
alpha = 0.05                        # significance level
rng = np.random.default_rng(42)     # fixed seed -> same random numbers on every run
results = []                        # every test adds one row here, shown in the summary at the end
print("Shape:", df.shape, "| Missing values:", int(df.isna().sum().sum()))
```

```python
# ---- pmf and cdf of a DISCRETE column: Customer_Rating (values 1 to 5) ----
counts = df['Customer_Rating'].value_counts().sort_index()
pmf = counts / len(df)               # pmf = count / N
cdf = pmf.cumsum()                   # cdf = running total of the pmf
print(pd.DataFrame({'count': counts, 'pmf': pmf.round(4), 'cdf': cdf.round(4)}))
print("\nP(rating <= 3) =", round(cdf[3], 4))
print("P(rating >= 4) =", round(1 - cdf[3], 4))

fig, axes = plt.subplots(1, 2, figsize=(10, 3.5))
axes[0].bar(pmf.index, pmf.values)                 # pmf as bars
axes[0].set_title('pmf of Customer_Rating')
axes[0].set_xlabel('Rating'); axes[0].set_ylabel('P(X = x)')
axes[1].step(cdf.index, cdf.values, where='post')  # cdf as steps
axes[1].set_title('cdf of Customer_Rating')
axes[1].set_xlabel('Rating'); axes[1].set_ylabel('P(X <= x)')
plt.tight_layout()
plt.show()
```

```python
# ---- pdf and cdf of a CONTINUOUS column: Fuel_Efficiency_kmpl ----
x = df['Fuel_Efficiency_kmpl'].to_numpy()
m, s = x.mean(), x.std()

def normal_pdf(v):
    # the normal curve written from its formula
    return 1 / (s * np.sqrt(2 * np.pi)) * np.exp(-(v - m) ** 2 / (2 * s ** 2))

grid = np.linspace(x.min(), x.max(), 200)
fig, axes = plt.subplots(1, 2, figsize=(11, 3.5))
axes[0].hist(x, bins=40, density=True, alpha=0.6)     # density=True -> bar areas add up to 1
axes[0].plot(grid, normal_pdf(grid), color='red')
axes[0].set_title('Histogram and normal pdf'); axes[0].set_xlabel('kmpl')
# Empirical cdf = share of values that are <= each value
sorted_x = np.sort(x)
ecdf = np.arange(1, len(x) + 1) / len(x)
axes[1].plot(sorted_x, ecdf, label='from data')
axes[1].plot(grid, stats.norm.cdf(grid, m, s), color='red', linestyle='--', label='normal cdf')
axes[1].set_title('cdf'); axes[1].set_xlabel('kmpl'); axes[1].legend()
plt.tight_layout()
plt.show()

# Probability that kmpl is between 15 and 20 = cdf(20) - cdf(15)
from_data = ((x >= 15) & (x <= 20)).mean()
from_normal = stats.norm.cdf(20, m, s) - stats.norm.cdf(15, m, s)
print("P(15 <= kmpl <= 20) from data  :", round(from_data, 4))
print("P(15 <= kmpl <= 20) from normal:", round(from_normal, 4))
```

```python
def skew(x):
    z = (x - x.mean()) / x.std()      # how many standard deviations away from the mean
    return (z ** 3).mean()            # average of the cubes: tail on the right -> positive

num_cols = ['Year', 'Mileage_KM', 'Engine_Displacement_CC', 'Fuel_Efficiency_kmpl',
            'Estimated_Repair_Time_Hours', 'Customer_Rating']
rows = []
for c in num_cols:
    x = df[c].to_numpy(dtype=float)
    sk = skew(x)
    if abs(sk) < 0.5:
        shape = 'symmetric'
    elif abs(sk) < 1:
        shape = 'moderately skewed'
    else:
        shape = 'highly skewed'
    rows.append({'column': c, 'mean': round(x.mean(), 2), 'median': round(np.median(x), 2),
                 'skewness': round(sk, 2), 'scipy skewness': round(stats.skew(x), 2), 'shape': shape})
pd.DataFrame(rows)
```

```python
pop = df['Estimated_Repair_Time_Hours'].to_numpy()      # the most skewed column
mu, sigma = pop.mean(), pop.std()
print("Population: mean = %.3f, sd = %.3f, skewness = %.2f" % (mu, sigma, skew(pop)))

rows = []
sample_means = {}
for n in [5, 30, 100, 500]:
    # take 2000 random samples of size n and keep the mean of each
    means = np.array([rng.choice(pop, size=n).mean() for i in range(2000)])
    sample_means[n] = means
    rows.append({'n': n,
                 'mean of sample means': round(means.mean(), 3),
                 'sd of sample means': round(means.std(), 3),
                 'theory sigma/sqrt(n)': round(sigma / np.sqrt(n), 3),
                 'skewness': round(skew(means), 2)})
print(pd.DataFrame(rows))

fig, axes = plt.subplots(1, 5, figsize=(16, 3))
axes[0].hist(pop, bins=40); axes[0].set_title('Column itself')
for ax, n in zip(axes[1:], [5, 30, 100, 500]):
    ax.hist(sample_means[n], bins=30); ax.set_title('Sample means, n = %d' % n)
plt.tight_layout()
plt.show()
```

```python
# Two questions:
# 1) H0: average Customer_Rating = 4.0   H1: it is not 4.0   (two-tailed)
# 2) H0: average Petrol Fuel_Efficiency = 16.0   H1: it is more than 16.0   (right-tailed)
rating = df['Customer_Rating'].to_numpy()
petrol = df.loc[df['Fuel'] == 'Petrol', 'Fuel_Efficiency_kmpl'].to_numpy()

z_crit_two = stats.norm.ppf(1 - alpha / 2)      # 1.96
z_crit_right = stats.norm.ppf(1 - alpha)        # 1.645
rows = []
for name, x, mu0, tail in [('Customer_Rating', rating, 4.0, 'two'), ('Petrol kmpl', petrol, 16.0, 'right')]:
    n = len(x)
    z = (x.mean() - mu0) / (x.std(ddof=1) / np.sqrt(n))     # the Z formula
    if tail == 'two':
        p = 2 * stats.norm.sf(abs(z))                       # both tails
        crit = z_crit_two
    else:
        p = stats.norm.sf(z)                                # right tail only
        crit = z_crit_right
    decision = 'reject H0' if p < alpha else 'fail to reject H0'
    rows.append({'test': name, 'mu0': mu0, 'sample mean': round(x.mean(), 3), 'n': n,
                 'Z': round(z, 3), 'critical': round(crit, 3), 'p-value': '%.3g' % p, 'decision': decision})
    results.append({'test': 'One-sample Z', 'what': name + ' mean = ' + str(mu0),
                    'statistic': round(z, 3), 'p-value': '%.3g' % p, 'decision': decision})
print(pd.DataFrame(rows).to_string(index=False))

# Library cross-check (one line): statsmodels ztest gives the same Z and p
print("\nLibrary check (rating):", sm.stats.ztest(rating, value=4.0))

# Graph: normal curve with the rejection area
v = np.linspace(-4, 4, 400)
plt.figure(figsize=(6, 3.2))
plt.plot(v, stats.norm.pdf(v))
plt.fill_between(v[v <= -z_crit_two], stats.norm.pdf(v[v <= -z_crit_two]), color='red', alpha=0.5)
plt.fill_between(v[v >= z_crit_two], stats.norm.pdf(v[v >= z_crit_two]), color='red', alpha=0.5)
plt.title('Two-tailed: reject H0 if |Z| > %.2f' % z_crit_two)
plt.show()
```

```python
# One-sample t test: same 2 questions as the Z test + a small sample of 12 Petrol records
small = pd.Series(petrol).sample(n=12, random_state=42).to_numpy()
rows = []
for name, x, mu0, tail in [('Customer_Rating', rating, 4.0, 'two'),
                           ('Petrol kmpl (all)', petrol, 16.0, 'right'),
                           ('Petrol kmpl (12 records)', small, 16.0, 'right')]:
    n = len(x)
    t = (x.mean() - mu0) / (x.std(ddof=1) / np.sqrt(n))     # the t formula
    dof = n - 1
    if tail == 'two':
        p = 2 * stats.t.sf(abs(t), dof)
        crit = stats.t.ppf(1 - alpha / 2, dof)
    else:
        p = stats.t.sf(t, dof)
        crit = stats.t.ppf(1 - alpha, dof)
    z_same = (x.mean() - mu0) / (x.std(ddof=1) / np.sqrt(n))
    p_z = 2 * stats.norm.sf(abs(z_same)) if tail == 'two' else stats.norm.sf(z_same)
    decision = 'reject H0' if p < alpha else 'fail to reject H0'
    rows.append({'test': name, 'n': n, 't': round(t, 3), 'df': dof, 'critical t': round(crit, 3),
                 'p (t curve)': '%.3g' % p, 'p (Z curve)': '%.3g' % p_z, 'decision': decision})
    results.append({'test': 'One-sample t', 'what': name + ' mean = ' + str(mu0),
                    'statistic': round(t, 3), 'p-value': '%.3g' % p, 'decision': decision})
print(pd.DataFrame(rows).to_string(index=False))

# Note: Proof. Library one-sample t test must give the same t and p (two-tailed rating case).
print("\nLibrary check (rating):", stats.ttest_1samp(rating, 4.0))
```

```python
# Two-sample t test (Welch)
diesel = df.loc[df['Fuel'] == 'Diesel', 'Fuel_Efficiency_kmpl'].to_numpy()
auto = df.loc[df['Transmission'] == 'Automatic', 'Mileage_KM'].to_numpy()
manual = df.loc[df['Transmission'] == 'Manual', 'Mileage_KM'].to_numpy()

rows = []
for name, a, b in [('kmpl: Diesel vs Petrol', diesel, petrol), ('Mileage_KM: Automatic vs Manual', auto, manual)]:
    se = np.sqrt(a.var(ddof=1) / len(a) + b.var(ddof=1) / len(b))     # standard error of the difference
    t = (a.mean() - b.mean()) / se
    dof = min(len(a), len(b)) - 1
    p = 2 * stats.t.sf(abs(t), dof)
    decision = 'reject H0' if p < alpha else 'fail to reject H0'
    t_lib, p_lib = stats.ttest_ind(a, b, equal_var=False)            # library cross-check (Welch)
    rows.append({'comparison': name, 'mean 1': round(a.mean(), 2), 'mean 2': round(b.mean(), 2),
                 't': round(t, 3), 'p-value': '%.3g' % p, 'library t': round(t_lib, 3), 'decision': decision})
    results.append({'test': 'Two-sample t', 'what': name, 'statistic': round(t, 3),
                    'p-value': '%.3g' % p, 'decision': decision})
print(pd.DataFrame(rows).to_string(index=False))
```

```python
diag_counts = df['Diagnosis'].value_counts().sort_index()
day = pd.to_datetime(df['Inspection_Date']).dt.day_name()
day_counts = day.value_counts().reindex(['Monday', 'Tuesday', 'Wednesday', 'Thursday', 'Friday', 'Saturday', 'Sunday'])

rows = []
for name, counts in [('Diagnosis (35 classes)', diag_counts), ('Inspection weekday (7 classes)', day_counts)]:
    O = counts.to_numpy(dtype=float)
    k = len(O)
    E = np.full(k, O.sum() / k)               # same expected count for every class
    chi = ((O - E) ** 2 / E).sum()            # chi-square formula
    dof = k - 1
    crit = stats.chi2.ppf(1 - alpha, dof)     # critical value
    p = stats.chi2.sf(chi, dof)               # p-value
    decision = 'reject H0' if p < alpha else 'fail to reject H0'
    lib = stats.chisquare(O)                  # library cross-check
    rows.append({'column': name, 'chi-square': round(chi, 2), 'df': dof, 'critical': round(crit, 2),
                 'p-value': '%.3g' % p, 'library chi-square': round(lib.statistic, 2), 'decision': decision})
    results.append({'test': 'Chi-square GOF', 'what': name + ' equally likely', 'statistic': round(chi, 2),
                    'p-value': '%.3g' % p, 'decision': decision})
print(pd.DataFrame(rows).to_string(index=False))

# Bar chart of Diagnosis counts against the equal share
sorted_counts = diag_counts.sort_values()
plt.figure(figsize=(12, 3.5))
plt.bar(range(35), sorted_counts.values)
plt.axhline(sorted_counts.mean(), color='red', linestyle='--', label='equal share = %.1f' % sorted_counts.mean())
plt.xlabel('Diagnosis (sorted by count)'); plt.ylabel('Records'); plt.legend()
plt.title('Diagnosis counts: smallest %d, largest %d' % (sorted_counts.min(), sorted_counts.max()))
plt.show()
```

```python
engine = df.loc[df['System_Category'] == 'Engine', 'Estimated_Repair_Time_Hours'].to_numpy()
brakes = df.loc[df['System_Category'] == 'Brakes', 'Estimated_Repair_Time_Hours'].to_numpy()

rows = []
for name, a, b in [('kmpl: Diesel vs Petrol', diesel, petrol),
                   ('Mileage_KM: Automatic vs Manual', auto, manual),
                   ('Repair time: Engine vs Brakes', engine, brakes)]:
    va, vb = a.var(ddof=1), b.var(ddof=1)
    if va >= vb:                                   # larger variance goes on top, so F >= 1
        F, df1, df2 = va / vb, len(a) - 1, len(b) - 1
    else:
        F, df1, df2 = vb / va, len(b) - 1, len(a) - 1
    crit = stats.f.ppf(1 - alpha / 2, df1, df2)    # upper 2.5% point
    p = min(1.0, 2 * stats.f.sf(F, df1, df2))
    decision = 'reject H0' if p < alpha else 'fail to reject H0'
    rows.append({'comparison': name, 'variance 1': round(va, 3), 'variance 2': round(vb, 3),
                 'F': round(F, 3), 'critical F': round(crit, 3), 'p-value': '%.3g' % p, 'decision': decision})
    results.append({'test': 'F test (2 variances)', 'what': name, 'statistic': round(F, 3),
                    'p-value': '%.3g' % p, 'decision': decision})
print(pd.DataFrame(rows).to_string(index=False))
```

```python
def one_way_anova(values, groups):
    values = np.asarray(values, dtype=float)
    groups = np.asarray(groups)
    grand = values.mean()                         # grand mean of all records
    ss_between = 0.0
    ss_within = 0.0
    names = np.unique(groups)
    for g in names:
        x = values[groups == g]
        ss_between += len(x) * (x.mean() - grand) ** 2     # group mean far from grand mean -> big
        ss_within += ((x - x.mean()) ** 2).sum()           # spread inside the group
    df_between = len(names) - 1
    df_within = len(values) - len(names)
    F = (ss_between / df_between) / (ss_within / df_within)
    p = stats.f.sf(F, df_between, df_within)
    return F, p, df_between, df_within, ss_between, ss_within

rows = []
for col, grp in [('Fuel_Efficiency_kmpl', 'Fuel'), ('Mileage_KM', 'Vehicle_Type')]:
    F, p, d1, d2, ssb, ssw = one_way_anova(df[col], df[grp])
    lib = stats.f_oneway(*[g[col].to_numpy() for _, g in df.groupby(grp)])      # library cross-check
    decision = 'reject H0' if p < alpha else 'fail to reject H0'
    rows.append({'column': col, 'grouped by': grp, 'F': round(F, 3), 'df': (d1, d2),
                 'critical F': round(stats.f.ppf(1 - alpha, d1, d2), 3), 'p-value': '%.3g' % p,
                 'library F': round(lib.statistic, 3), 'decision': decision})
    results.append({'test': 'One-way ANOVA', 'what': col + ' by ' + grp, 'statistic': round(F, 3),
                    'p-value': '%.3g' % p, 'decision': decision})
    print(col, 'by', grp)
    print(df.groupby(grp)[col].agg(['count', 'mean', 'std']).round(2), "\n")
print(pd.DataFrame(rows).to_string(index=False))
```

```python
y, a, b = 'Fuel_Efficiency_kmpl', 'Fuel', 'Transmission'

# Balanced sample: same number of records in each cell (the smallest cell decides)
cell_size = df.groupby([a, b]).size()
n = int(cell_size.min())
bal = df.groupby([a, b], group_keys=False).sample(n=n, random_state=42)
print("Records per cell:", n, "| total:", len(bal))

ka, kb = bal[a].nunique(), bal[b].nunique()      # number of levels of each factor
G = bal[y].mean()                                # grand mean
mean_a = bal.groupby(a)[y].mean()
mean_b = bal.groupby(b)[y].mean()
cell_mean = bal.groupby([a, b])[y].mean()

ss_a = kb * n * ((mean_a - G) ** 2).sum()
ss_b = ka * n * ((mean_b - G) ** 2).sum()
ss_cells = n * ((cell_mean - G) ** 2).sum()
ss_ab = ss_cells - ss_a - ss_b                   # what the two factors alone cannot explain = interaction
ss_e = ((bal[y] - bal.groupby([a, b])[y].transform('mean')) ** 2).sum()

df_a, df_b, df_ab, df_e = ka - 1, kb - 1, (ka - 1) * (kb - 1), len(bal) - ka * kb
ms_e = ss_e / df_e
table = []
for name, ss, d in [(a, ss_a, df_a), (b, ss_b, df_b), (a + ' x ' + b, ss_ab, df_ab)]:
    F = (ss / d) / ms_e
    p = stats.f.sf(F, d, df_e)
    decision = 'reject H0' if p < alpha else 'fail to reject H0'
    table.append({'effect': name, 'SS': round(ss, 2), 'df': d, 'F': round(F, 3), 'p-value': '%.3g' % p, 'decision': decision})
    results.append({'test': 'Two-way ANOVA', 'what': y + ': ' + name, 'statistic': round(F, 3),
                    'p-value': '%.3g' % p, 'decision': decision})
table.append({'effect': 'Error', 'SS': round(ss_e, 2), 'df': df_e, 'F': '', 'p-value': '', 'decision': ''})
print(pd.DataFrame(table).to_string(index=False))

# Library cross-check: statsmodels must give the same F values
model = smf.ols('Fuel_Efficiency_kmpl ~ C(Fuel) * C(Transmission)', data=bal).fit()
print("\nLibrary check:")
print(sm.stats.anova_lm(model, typ=2).round(3))
```

```python
# Interaction plot: if the two lines are not parallel, there is an interaction
means = bal.pivot_table(index='Fuel', columns='Transmission', values=y, aggfunc='mean')
print(means.round(2))
plt.figure(figsize=(5.5, 3.5))
for t in means.columns:
    plt.plot(means.index, means[t], marker='o', label=t)
plt.ylabel('Mean ' + y); plt.title('Fuel x Transmission'); plt.legend()
plt.show()
```

# EXPERIMENT 5:

```python
import numpy as np                 # numbers (used for the correlation formula)
import pandas as pd                # tables and group calculations
import matplotlib.pyplot as plt    # all graphs of this experiment

df = pd.read_csv('vehicle_diagnostic_clean.csv', parse_dates=['Inspection_Date'])
severity_order = ['Low', 'Medium', 'High', 'Critical']      # natural order of the severity levels
print("Shape:", df.shape, "| Missing values:", int(df.isna().sum().sum()))
```

```python
# (a) Number of records in each category of four columns
cat_cols = ['System_Category', 'Severity_Level', 'Fuel', 'Vehicle_Type']
fig, axes = plt.subplots(2, 2, figsize=(12, 7))
for ax, c in zip(axes.ravel(), cat_cols):
    counts = df[c].value_counts()                  # records of each category, largest first
    if c == 'Severity_Level':
        counts = counts.reindex(severity_order)    # show Low -> Critical, not by size
    bars = ax.bar(counts.index, counts.values)
    ax.bar_label(bars)                             # write the count on top of each bar
    ax.set_title('Records by ' + c)
    ax.tick_params(axis='x', rotation=25)
plt.tight_layout()
plt.show()

for c in cat_cols:
    print(c, (df[c].value_counts(normalize=True) * 100).round(1).to_dict())
```

```python
# (b) Records of each Diagnosis (the target column of the project), as horizontal bars
diag = df['Diagnosis'].value_counts().sort_values()
equal_share = len(df) / df['Diagnosis'].nunique()      # records per class if all classes were equal
plt.figure(figsize=(9, 9))
plt.barh(diag.index, diag.values)
plt.axvline(equal_share, color='red', linestyle='--', label='equal share = %.1f' % equal_share)
plt.title('Records of each Diagnosis')
plt.xlabel('Records')
plt.legend()
plt.tight_layout()
plt.show()
print("Smallest:", diag.index[0], diag.iloc[0], "| Largest:", diag.index[-1], diag.iloc[-1])
print("Diagnoses below the equal share:", int((diag < equal_share).sum()), "of", len(diag))
```

```python
def box_numbers(x):
    # the numbers behind one box of a box plot
    q1, median, q3 = np.percentile(x, [25, 50, 75])
    iqr = q3 - q1
    outliers = ((x < q1 - 1.5 * iqr) | (x > q3 + 1.5 * iqr)).sum()
    return 'median %.2f, Q1 %.2f, Q3 %.2f, outliers %d' % (median, q1, q3, outliers)

# (a) Fuel efficiency by Fuel
fuels = ['Petrol', 'Diesel', 'CNG']
groups = [df.loc[df['Fuel'] == f, 'Fuel_Efficiency_kmpl'] for f in fuels]
plt.figure(figsize=(6, 3.5))
plt.boxplot(groups, tick_labels=fuels)
plt.title('Fuel_Efficiency_kmpl by Fuel')
plt.show()
for f, g in zip(fuels, groups):
    print(f, ':', box_numbers(g))
```

```python
def heatmap(table, title, cmap):
    plt.figure(figsize=(8, 4.5))
    im = plt.imshow(table.values, cmap=cmap)               # colour every cell by its value
    plt.colorbar()
    plt.xticks(range(table.shape[1]), table.columns, rotation=45, ha='right')
    plt.yticks(range(table.shape[0]), table.index)
    for i in range(table.shape[0]):
        for j in range(table.shape[1]):
            r, g, b, a = im.cmap(im.norm(table.values[i, j]))          # colour of this cell
            text_colour = 'white' if 0.299 * r + 0.587 * g + 0.114 * b < 0.5 else 'black'   # white text on a dark cell
            plt.text(j, i, '%.2f' % table.values[i, j], ha='center', va='center', fontsize=8, color=text_colour)
    plt.title(title)
    plt.tight_layout()
    plt.show()

# (a) Correlation by hand: standardise every column, then r = (Z^T x Z) / n
num_cols = ['Year', 'Mileage_KM', 'Engine_Displacement_CC', 'Fuel_Efficiency_kmpl',
            'Estimated_Repair_Time_Hours', 'Customer_Rating']
Z = (df[num_cols] - df[num_cols].mean()) / df[num_cols].std(ddof=0)
corr = pd.DataFrame(Z.T.values @ Z.values / len(df), index=num_cols, columns=num_cols)
heatmap(corr, 'Correlation of numeric columns', 'coolwarm')

# Note: Proof. Our hand calculation must match pandas .corr()
print("Largest difference from pandas corr():", (corr - df[num_cols].corr()).abs().max().max())
print("r(Year, Mileage_KM)                = %.2f" % corr.loc['Year', 'Mileage_KM'])
print("r(Engine_CC, Fuel_Efficiency_kmpl) = %.2f" % corr.loc['Engine_Displacement_CC', 'Fuel_Efficiency_kmpl'])
```

```python
# (b) Which System_Category fails for which Fuel / Vehicle_Type? (row percentages)
overall = df['System_Category'].value_counts(normalize=True) * 100
for column in ['Fuel', 'Vehicle_Type']:
    counts = pd.crosstab(df[column], df['System_Category'])
    percent = counts.div(counts.sum(axis=1), axis=0) * 100         # each row adds up to 100
    heatmap(percent, 'System_Category share (%) within each ' + column, 'Blues')
    gap = (percent - overall).abs().max().max()
    print(column, ": largest gap from the overall share = %.1f points" % gap)
```

# EXPERIMENT 8:
# ASSOCIATION RULE MINING (Apriori and FP-Growth)

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from itertools import combinations      # to make groups of items

df = pd.read_csv('vehicle_diagnostic_clean.csv')
pd.set_option('display.width', 250)
pd.set_option('display.max_colwidth', 60)

# Step 0: turn every row into a "basket" (transaction):
# the symptoms of that vehicle + its diagnosis (written with the prefix 'DX: ')
symptom_cols = ['Symptom_1', 'Symptom_2', 'Symptom_3']
transactions = []
for i in range(len(df)):
    items = set(df.loc[i, symptom_cols]) - {'No_Symptom'}     # remove the filler 'No_Symptom'
    items.add('DX: ' + df.loc[i, 'Diagnosis'])
    transactions.append(items)

N = len(transactions)
print("Number of transactions:", N)
print("Different items       :", len(set().union(*transactions)))
print("Average items per transaction: %.2f" % np.mean([len(t) for t in transactions]))
for t in transactions[:3]:
    print(sorted(t))

# Note: 169 items = 134 symptoms + 35 diagnoses. 'No_Symptom' is removed because an empty slot is not a symptom
# (otherwise it would become the most frequent "item" without any meaning).
```

```python
min_support = 0.005
min_count = int(np.ceil(min_support * N))
print("Minimum count:", min_count)

# Level 1: count every single item
counts = {}
for t in transactions:
    for item in t:
        counts[(item,)] = counts.get((item,), 0) + 1
level = {k: v for k, v in counts.items() if v >= min_count}     # frequent itemsets of the current size
frequent = dict(level)                                          # all frequent itemsets of all sizes
print("Size 1: frequent =", len(level))

size = 1
while len(level) > 0:
    # (a) Join: two itemsets that share all items except the last one make a bigger candidate
    keys = sorted(level)
    candidates = set()
    for a in range(len(keys)):
        for b in range(a + 1, len(keys)):
            if keys[a][:-1] != keys[b][:-1]:
                break                                       # keys are sorted, so no later one can match
            new = keys[a] + (keys[b][-1],)
            # (b) Prune: every smaller subset of the candidate must be frequent
            if all(sub in level for sub in combinations(new, size)):
                candidates.add(new)
    size += 1
    # (c) Count the candidates in the data
    counts = {}
    for t in transactions:
        for group in combinations(sorted(t), size):
            if group in candidates:
                counts[group] = counts.get(group, 0) + 1
    level = {k: v for k, v in counts.items() if v >= min_count}
    frequent.update(level)
    print("Size %d: candidates = %d, frequent = %d" % (size, len(candidates), len(level)))

print("\nTotal frequent itemsets:", len(frequent))

# Note: Proof. Count the top 2-item itemset directly in the table with pandas.
top = max([k for k in frequent if len(k) == 2], key=lambda k: frequent[k])
print("Most frequent pair:", top, "-> Apriori count", frequent[top])
check = pd.Series(True, index=df.index)
for item in top:
    if item.startswith('DX: '):
        check &= df['Diagnosis'] == item[4:]
    else:
        check &= df[symptom_cols].eq(item).any(axis=1)
print("Direct pandas count:", check.sum())
```

```python
rule_rows = []
for itemset, count_xy in frequent.items():
    if len(itemset) < 2:
        continue
    for item in itemset:
        if not item.startswith('DX: '):
            continue                                            # the "then" part must be a diagnosis
        left = tuple(i for i in itemset if i != item)           # the symptoms
        if any(i.startswith('DX: ') for i in left):
            continue                                            # the "if" part must have only symptoms
        count_x = frequent[left]                                # records with these symptoms
        count_y = frequent[(item,)]                             # records with this diagnosis
        confidence = count_xy / count_x
        lift = confidence / (count_y / N)
        rule_rows.append({'if (symptoms)': ' + '.join(left), 'then (diagnosis)': item[4:],
                          'count': count_xy, 'support': count_xy / N,
                          'confidence': confidence, 'lift': lift})
rules = pd.DataFrame(rule_rows)

n_two = rules['if (symptoms)'].str.contains(' + ', regex=False).sum()
print("Number of rules:", len(rules), "| with 1 symptom:", len(rules) - n_two, "| with 2 symptoms:", n_two)
print("Rules with confidence = 1.0 :", (rules['confidence'] == 1).sum())
print("Rules with confidence >= 0.8:", (rules['confidence'] >= 0.8).sum())
print("\nStrongest rules:")
print(rules.sort_values(['confidence', 'count'], ascending=False).head(8).round(3).to_string(index=False))

# Note: Proof. Calculate the confidence of the most frequent rule directly with pandas.
r = rules.sort_values('count', ascending=False).iloc[0]
has_symptom = df[symptom_cols].eq(r['if (symptoms)']).any(axis=1)
both = has_symptom & (df['Diagnosis'] == r['then (diagnosis)'])
print("\n", r['if (symptoms)'], "->", r['then (diagnosis)'])
print("records with the symptom:", has_symptom.sum(), "| of them with the diagnosis:", both.sum(),
      "| confidence = %.3f (rule table: %.3f)" % (both.sum() / has_symptom.sum(), r['confidence']))
```

```python
# How useful are the strong rules (confidence >= 0.8) as a diagnosis tool?
strong = rules[rules['confidence'] >= 0.8].sort_values('confidence', ascending=False)
covered = 0         # records where at least one strong rule applies
correct = 0         # of those, records where the first matching rule gives the right diagnosis
for t in transactions:
    symptoms_of_vehicle = {i for i in t if not i.startswith('DX: ')}
    for left, diagnosis in zip(strong['if (symptoms)'], strong['then (diagnosis)']):
        if set(left.split(' + ')) <= symptoms_of_vehicle:       # all symptoms of the rule are present
            covered += 1
            if ('DX: ' + diagnosis) in t:
                correct += 1
            break
print("Records covered by a strong rule: %d of %d (%.1f%%)" % (covered, N, 100 * covered / N))
print("Diagnoses that have a strong rule: %d of 35" % strong['then (diagnosis)'].nunique())
print("Rule gives the right diagnosis: %d of %d (%.1f%%)" % (correct, covered, 100 * correct / covered))
# This is measured on the same data the rules came from, so it is an optimistic number.
```

```python
class Node:
    # one box of the tree: an item, how many transactions pass through it, its parent and children
    def __init__(self, item, parent):
        self.item = item
        self.count = 0
        self.parent = parent
        self.children = {}

def build_tree(paths, min_count):
    # paths = list of (list of items, how many times this path occurs)
    total = {}
    for items, c in paths:
        for item in items:
            total[item] = total.get(item, 0) + c
    keep = {i: c for i, c in total.items() if c >= min_count}      # frequent items only
    root = Node(None, None)
    header = {}                                                    # item -> all tree nodes of this item
    for items, c in paths:
        items = [i for i in items if i in keep]
        items.sort(key=lambda i: (-keep[i], i))                    # most frequent item first
        node = root
        for item in items:
            if item not in node.children:
                child = Node(item, node)
                node.children[item] = child
                header.setdefault(item, []).append(child)
            node = node.children[item]
            node.count += c
    return header, keep

def mine(header, keep, suffix, result):
    # for each item: it is a frequent itemset together with the suffix; then build a smaller tree from the paths above it
    for item in keep:
        pattern = tuple(sorted(suffix + (item,)))
        result[pattern] = keep[item]
        small_paths = []
        for node in header[item]:
            up = []
            parent = node.parent
            while parent.item is not None:                         # walk up to the root
                up.append(parent.item)
                parent = parent.parent
            if len(up) > 0:
                small_paths.append((up, node.count))
        if len(small_paths) > 0:
            header2, keep2 = build_tree(small_paths, min_count)
            mine(header2, keep2, suffix + (item,), result)

header, keep = build_tree([(list(t), 1) for t in transactions], min_count)
fp_frequent = {}
mine(header, keep, (), fp_frequent)

print("Itemsets found by Apriori   :", len(frequent))
print("Itemsets found by FP-Growth :", len(fp_frequent))

# Note: Proof. Both algorithms must give the same itemsets with the same counts.
print("Same itemsets?", set(frequent) == set(fp_frequent))
print("Itemsets with a different count:", sum(1 for k in frequent if frequent[k] != fp_frequent.get(k)))
```

# EXPERIMENT 6:
# CLUSTERING (K-Means, Hierarchical, DBSCAN, GMM)

```python
cluster_cols = ['Year', 'Mileage_KM', 'Engine_Displacement_CC', 'Fuel_Efficiency_kmpl']
X = df[cluster_cols].to_numpy(dtype=float)

# Scaling (z-score): distance based methods need all columns on the same scale.
# Without it, Mileage_KM (about 100000) would decide everything and Year (about 2017) would not matter.
col_mean = X.mean(axis=0)
col_std = X.std(axis=0)
Z = (X - col_mean) / col_std

print("Std before scaling:", X.std(axis=0).round(1))
print("Mean after scaling:", Z.mean(axis=0).round(6))
print("Std after scaling :", Z.std(axis=0).round(6))
```

```python
def kmeans(Z, k, seed, max_steps=100):
    rng = np.random.default_rng(seed)
    centers = Z[rng.choice(len(Z), size=k, replace=False)]      # k random vehicles as the first centres
    for step in range(max_steps):
        # squared distance of every vehicle to every centre
        dist = ((Z[:, None, :] - centers[None, :, :]) ** 2).sum(axis=2)
        labels = dist.argmin(axis=1)                            # nearest centre
        new_centers = np.array([Z[labels == j].mean(axis=0) for j in range(k)])
        if np.allclose(new_centers, centers):                   # centres did not move -> done
            break
        centers = new_centers
    inertia = ((Z - centers[labels]) ** 2).sum()
    return labels, centers, inertia

# Best of 5 random starts for every k from 1 to 8
best = {}
for k in range(1, 9):
    runs = [kmeans(Z, k, seed) for seed in range(5)]
    best[k] = min(runs, key=lambda r: r[2])

# Elbow: inertia against k. We look for the point after which the drop becomes small.
inertias = [best[k][2] for k in range(1, 9)]
print(pd.DataFrame({'k': range(1, 9), 'inertia': np.round(inertias, 1)}).to_string(index=False))
plt.figure(figsize=(6, 3.5))
plt.plot(range(1, 9), inertias, marker='o')
plt.xlabel('k'); plt.ylabel('Inertia'); plt.title('Elbow graph')
plt.show()
```

```python
# Silhouette: for every vehicle, a = average distance to its own group, b = average distance to the nearest other group.
# s = (b - a) / max(a, b). Close to 1 = well placed, near 0 = on the border, negative = maybe in the wrong group.
def silhouette(Z, labels):
    k = labels.max() + 1
    sizes = np.bincount(labels, minlength=k)
    scores = np.zeros(len(Z))
    for start in range(0, len(Z), 500):                          # 500 vehicles at a time (saves memory)
        part = Z[start:start + 500]
        dist = np.sqrt(((part[:, None, :] - Z[None, :, :]) ** 2).sum(axis=2))
        avg = np.zeros((len(part), k))                           # average distance to every group
        for j in range(k):
            avg[:, j] = dist[:, labels == j].sum(axis=1) / sizes[j]
        own = labels[start:start + 500]
        rows = np.arange(len(part))
        a = avg[rows, own] * sizes[own] / (sizes[own] - 1)       # own group: leave the vehicle itself out
        avg[rows, own] = np.inf
        b = avg.min(axis=1)
        scores[start:start + 500] = (b - a) / np.maximum(a, b)
    return scores

sil = {}
for k in range(2, 9):
    sil[k] = silhouette(Z, best[k][0]).mean()
print(pd.DataFrame({'k': list(sil), 'average silhouette': np.round(list(sil.values()), 3)}).to_string(index=False))

# Note: Proof. Library silhouette must give the same average for k = 3.
from sklearn.metrics import silhouette_score
print("Our k=3: %.4f | library k=3: %.4f" % (sil[3], silhouette_score(Z, best[3][0])))
```

```python
# We choose k = 3 (the elbow bends at 3 and the silhouette is good enough there)
labels, centers, inertia = best[3]
total = ((Z - Z.mean(axis=0)) ** 2).sum()           # inertia with k = 1
print("Inertia k=3: %.1f | total (k=1): %.1f | share explained: %.1f%%" % (inertia, total, 100 * (1 - inertia / total)))

# Note: Proof. Library K-Means inertia (almost the same).
from sklearn.cluster import KMeans
print("Library inertia k=3: %.1f" % KMeans(n_clusters=3, n_init=10, random_state=0).fit(Z).inertia_)

# Give the groups a fixed order and name by looking at their average values (0 = newest cars)
order = np.argsort(-centers[:, 0])
new_id = np.empty(3, dtype=int)
new_id[order] = np.arange(3)
df['Group'] = new_id[labels]
names = {0: 'Newer, low mileage', 1: 'Big engine', 2: 'Older, high mileage'}
df['Group_Name'] = df['Group'].map(names)
profile = df.groupby('Group_Name')[cluster_cols].mean().round(1).loc[list(names.values())]
profile['records'] = df['Group_Name'].value_counts()[list(names.values())]
print(profile)

# Scatter plot of the groups
plt.figure(figsize=(6, 4))
for g in range(3):
    m = df['Group'] == g
    plt.scatter(df.loc[m, 'Year'], df.loc[m, 'Mileage_KM'], s=4, label=names[g])
plt.xlabel('Year'); plt.ylabel('Mileage_KM'); plt.legend(markerscale=3)
plt.title('K-Means groups (k = 3)')
plt.show()
```

```python
# Do the groups mean anything? Compare how much the Fuel (etc.) mix changes between groups
# with the same measure for RANDOM groups of the same sizes. Only a clearly bigger number is a real pattern.
def mix_difference(group, column):
    overall = df[column].value_counts(normalize=True)
    total = 0.0
    for g in np.unique(group):
        share = df.loc[group == g, column].value_counts(normalize=True).reindex(overall.index).fillna(0)
        total += (group == g).mean() * 0.5 * (share - overall).abs().sum()    # weighted difference from overall mix
    return total

rng = np.random.default_rng(0)
group_array = df['Group'].to_numpy()
rows = []
for column in ['Fuel', 'Vehicle_Type', 'Severity_Level', 'System_Category', 'Diagnosis']:
    real = mix_difference(group_array, column)
    random_ones = [mix_difference(rng.permutation(group_array), column) for i in range(100)]
    rows.append({'column': column, 'real groups': round(real, 3),
                 'random groups (average)': round(np.mean(random_ones), 3),
                 'random groups (largest of 100)': round(max(random_ones), 3)})
print(pd.DataFrame(rows).to_string(index=False))
```

```python
rng = np.random.default_rng(0)
sample_rows = rng.choice(len(Z), size=200, replace=False)
S = Z[sample_rows]

def hierarchical(S, method):
    n = len(S)
    D = np.sqrt(((S[:, None, :] - S[None, :, :]) ** 2).sum(axis=2))    # distance table
    np.fill_diagonal(D, np.inf)                                        # a group is not joined with itself
    size = np.ones(n)                                                  # members in each group
    name = np.arange(n)                                                # current name of each group
    merges = []
    for step in range(n - 1):
        i, j = np.unravel_index(D.argmin(), D.shape)                   # the two closest groups
        height = D[i, j]
        if method == 'single':
            new_row = np.minimum(D[i], D[j])                           # closest pair
        else:
            new_row = (size[i] * D[i] + size[j] * D[j]) / (size[i] + size[j])    # average linkage update
        D[i, :] = new_row                                              # group i becomes the merged group
        D[:, i] = new_row
        D[i, i] = np.inf
        D[j, :] = np.inf                                               # group j is gone
        D[:, j] = np.inf
        merges.append([min(name[i], name[j]), max(name[i], name[j]), height, size[i] + size[j]])
        size[i] = size[i] + size[j]
        name[i] = n + step
    return np.array(merges)

def cut(merges, n, k):
    # undo the last k-1 merges to get k groups
    members = {i: [i] for i in range(n)}
    for step in range(n - k):
        a, b = int(merges[step, 0]), int(merges[step, 1])
        members[n + step] = members.pop(a) + members.pop(b)
    result = np.empty(n, dtype=int)
    for g, m in enumerate(members.values()):
        result[m] = g
    return result

merges = hierarchical(S, 'average')
print("Last 3 merges (cluster A, cluster B, distance, size):")
print(merges[-3:].round(3))

# Note: Proof. scipy linkage must give the same merge distances.
from scipy.cluster.hierarchy import linkage
print("Largest difference in merge distances from scipy:", np.abs(merges[:, 2] - linkage(S, 'average')[:, 2]).max())

# Merge distance of the last 15 merges: a big jump means two very different groups are being joined
plt.figure(figsize=(6, 3.5))
plt.plot(range(1, 16), merges[-15:, 2], marker='o')
plt.xlabel('Merge number (last 15 merges)'); plt.ylabel('Distance'); plt.title('Merge distances')
plt.show()
```

```python
# Cut into 3 groups: average linkage versus single linkage
for method in ['single', 'average']:
    groups = cut(hierarchical(S, method), len(S), 3)
    print(method, "group sizes:", sorted(np.bincount(groups).tolist(), reverse=True))

# Compare the average-linkage groups with the K-Means groups of the same 200 vehicles
hier = cut(merges, len(S), 3)
kmeans_groups = df['Group'].to_numpy()[sample_rows]
table = pd.crosstab(pd.Series(hier, name='hierarchical'), pd.Series(kmeans_groups, name='K-Means'))
print(table)
print("Agreement with K-Means: %.1f%%" % (100 * table.max(axis=1).sum() / len(S)))
```

```python
min_pts = 8

# Distance of every vehicle to its 8th nearest vehicle (the vehicle itself is the 1st, at distance 0)
kth = np.zeros(len(Z))
for start in range(0, len(Z), 500):
    part = Z[start:start + 500]
    dist = np.sqrt(((part[:, None, :] - Z[None, :, :]) ** 2).sum(axis=2))
    dist.sort(axis=1)
    kth[start:start + 500] = dist[:, min_pts - 1]
k_distance = np.sort(kth)
print("k-distance percentiles (10, 50, 90, 99, 100):", np.percentile(k_distance, [10, 50, 90, 99, 100]).round(3))
plt.figure(figsize=(6, 3.5))
plt.plot(k_distance)
plt.xlabel('Vehicles (sorted)'); plt.ylabel('Distance to 8th nearest vehicle'); plt.title('k-distance graph')
plt.show()
```

```python
def dbscan(Z, eps, min_pts):
    n = len(Z)
    # neighbours of every vehicle = vehicles within distance eps
    neighbors = []
    for start in range(0, n, 500):
        part = Z[start:start + 500]
        dist = np.sqrt(((part[:, None, :] - Z[None, :, :]) ** 2).sum(axis=2))
        for r in range(len(part)):
            neighbors.append(np.where(dist[r] <= eps)[0])
    is_core = np.array([len(nb) >= min_pts for nb in neighbors])
    labels = np.full(n, -1)                  # -1 = noise until a group takes the vehicle
    group = 0
    for i in range(n):
        if labels[i] != -1 or not is_core[i]:
            continue
        labels[i] = group                    # start a new group from an unused core point
        queue = [i]
        while len(queue) > 0:
            p = queue.pop()
            if not is_core[p]:
                continue                     # border points join the group but do not spread it
            for q in neighbors[p]:
                if labels[q] == -1:
                    labels[q] = group
                    queue.append(q)
        group += 1
    return labels, is_core

# Try several eps values
for eps in [0.3, 0.4, 0.5, 0.6, 0.7]:
    labels_try, core_try = dbscan(Z, eps, min_pts)
    sizes = sorted(np.bincount(labels_try[labels_try >= 0]).tolist(), reverse=True)
    print("eps %.1f: groups = %d, biggest sizes = %s, noise = %d" % (eps, labels_try.max() + 1, sizes[:4], (labels_try == -1).sum()))
```

```python
# We choose eps = 0.6: smaller values break the data into many small groups, 0.6 gives two clear groups
labels_db, is_core = dbscan(Z, 0.6, min_pts)
df['DBSCAN'] = labels_db
print("Group sizes (-1 = noise):")
print(df['DBSCAN'].value_counts().sort_index().to_string())
print("Core points:", is_core.sum(), "| border points:", ((labels_db >= 0) & ~is_core).sum())
print("\nAverage of the 4 columns for each DBSCAN group:")
print(df.groupby('DBSCAN')[cluster_cols].mean().round(1))

# Note: Proof. The library DBSCAN must give exactly the same labels.
from sklearn.cluster import DBSCAN
library_labels = DBSCAN(eps=0.6, min_samples=min_pts).fit(Z).labels_
print("\nSame labels as the library?", (library_labels == labels_db).all())
```

```python
def log_gaussian(Z, mean, cov):
    # log of the Gaussian density of every row of Z
    d = Z.shape[1]
    diff = Z - mean
    sign, logdet = np.linalg.slogdet(cov)
    solved = np.linalg.solve(cov, diff.T).T
    mahal = (diff * solved).sum(axis=1)                      # squared distance that accounts for the spread
    return -0.5 * (d * np.log(2 * np.pi) + logdet + mahal)

def gmm(Z, k, start_means, reg=0.01, max_steps=500, tol=1e-6):
    n, d = Z.shape
    weights = np.full(k, 1 / k)
    means = start_means.copy()
    covs = np.array([np.eye(d) for j in range(k)])
    old_ll = -np.inf
    for step in range(max_steps):
        # E-step: probability of every group for every vehicle
        log_p = np.stack([np.log(weights[j]) + log_gaussian(Z, means[j], covs[j]) for j in range(k)], axis=1)
        top = log_p.max(axis=1, keepdims=True)
        log_total = top[:, 0] + np.log(np.exp(log_p - top).sum(axis=1))       # (log-sum-exp keeps numbers safe)
        resp = np.exp(log_p - log_total[:, None])
        ll = log_total.sum()
        # M-step: new weights, means and spreads
        nk = resp.sum(axis=0)
        weights = nk / n
        means = (resp.T @ Z) / nk[:, None]
        for j in range(k):
            diff = Z - means[j]
            covs[j] = (resp[:, j, None] * diff).T @ diff / nk[j] + reg * np.eye(d)
        if abs(ll - old_ll) < tol:
            break
        old_ll = ll
    return weights, means, covs, resp, ll

# Start from the K-Means centres (k = 3)
start = best[3][1]
weights, means, covs, resp, ll = gmm(Z, 3, start)
print("Log-likelihood: %.1f" % ll)
profile = pd.DataFrame(means * col_std + col_mean, columns=cluster_cols).round(1)
profile['weight'] = weights.round(3)
print(profile)

# Note: Proof. The library GMM with the same start and settings must give the same log-likelihood.
from sklearn.mixture import GaussianMixture
lib = GaussianMixture(n_components=3, covariance_type='full', weights_init=np.full(3, 1 / 3), means_init=start,
                      precisions_init=np.array([np.eye(4)] * 3), reg_covar=0.01, tol=1e-10, max_iter=2000).fit(Z)
print("Library log-likelihood: %.1f" % (lib.score(Z) * len(Z)))
```

```python
# Soft membership: how sure is the model about each vehicle?
sure = resp.max(axis=1)
print("Vehicles whose best probability is below 0.9: %.1f%%" % (100 * (sure < 0.9).mean()))
print("Vehicles whose best probability is below 0.6: %.1f%%" % (100 * (sure < 0.6).mean()))
print("Probabilities of the first 3 vehicles:")
print(resp[:3].round(3))

# Can BIC choose k? BIC = (number of parameters) x ln(n) - 2 x log-likelihood. Lower is better.
rows = []
for k in range(1, 9):
    out = gmm(Z, k, best[k][1])
    n_par = k * (4 + 4 * 5 // 2) + (k - 1)          # means + covariance entries + weights
    rows.append({'k': k, 'log-likelihood': round(out[4], 1), 'BIC': round(n_par * np.log(len(Z)) - 2 * out[4], 1)})
print(pd.DataFrame(rows).to_string(index=False))
```

```python
def sq_dist(A):
    # squared distance between every pair of rows
    s = (A ** 2).sum(axis=1)
    return np.maximum(s[:, None] + s[None, :] - 2 * A @ A.T, 0)

def affinities(A, perplexity=30):
    n = len(A)
    D = sq_dist(A)
    P = np.zeros((n, n))
    target = np.log(perplexity)
    reached = np.zeros(n)
    for i in range(n):
        d = np.delete(D[i], i)                          # distances to all other vehicles
        low, high, beta = 0.0, np.inf, 1.0              # beta = 1 / (2 x spread^2), found by binary search
        for step in range(50):
            p = np.exp(-(d - d.min()) * beta)
            p = p / p.sum()
            entropy = -(p * np.log(p + 1e-12)).sum()
            if abs(entropy - target) < 1e-5:
                break
            if entropy > target:                        # too many neighbours -> make the curve narrower
                low = beta
                beta = beta * 2 if high == np.inf else (beta + high) / 2
            else:                                       # too few neighbours -> make it wider
                high = beta
                beta = (beta + low) / 2
        P[i, np.arange(n) != i] = p
        reached[i] = np.exp(entropy)
    P = (P + P.T) / (2 * n)                             # make it symmetric, total = 1
    return P, reached

def tsne(A, perplexity=30, steps=1500, lr=50, seed=0):
    n = len(A)
    P, reached = affinities(A, perplexity)
    rng = np.random.default_rng(seed)
    Y = rng.normal(0, 1e-4, size=(n, 2))                # random start in 2D
    speed = np.zeros_like(Y)
    for it in range(steps):
        exaggeration = 12 if it < 250 else 1            # first steps: stronger pull, so groups form early
        momentum = 0.5 if it < 250 else 0.8
        num = 1 / (1 + sq_dist(Y))
        np.fill_diagonal(num, 0)
        Q = num / num.sum()
        W = (exaggeration * P - Q) * num
        grad = 4 * (W.sum(axis=1)[:, None] * Y - W @ Y)   # gradient of KL
        speed = momentum * speed - lr * grad
        Y = Y + speed
        Y = Y - Y.mean(axis=0)
    kl = (P * np.log((P + 1e-12) / (Q + 1e-12))).sum()
    return Y, kl, P, reached

# Note: Proof for the p values: total must be 1, matrix symmetric, perplexity must reach 30.
P_test, reached = affinities(A_symptom, 30)
print("Sum of all p: %.4f | symmetric: %s" % (P_test.sum(), np.allclose(P_test, P_test.T)))
print("Perplexity reached: smallest %.3f, largest %.3f (target 30)" % (reached.min(), reached.max()))
```

```python
blocks = {}      # name of a feature group -> (matrix for all rows, column names)

# 1) Symptoms (multi-hot)
symptom_names = sorted(set(df[symptom_cols].to_numpy().ravel()) - {'No_Symptom'})
M_sym = np.zeros((len(df), len(symptom_names)))
for r in range(len(df)):
    for s in df.loc[r, symptom_cols]:
        if s != 'No_Symptom':
            M_sym[r, symptom_names.index(s)] = 1
blocks['Symptoms'] = (M_sym, symptom_names)

# 2) Follow-up tokens (question | answer), multi-hot
tokens = []
for r in range(len(df)):
    t = []
    for q, a in [('Followup_Question_1', 'Followup_Answer_1'), ('Followup_Question_2', 'Followup_Answer_2')]:
        if df.loc[r, q] != 'No_Followup':
            t.append(df.loc[r, q] + ' | ' + df.loc[r, a])
    tokens.append(t)
token_names = sorted(set(t for row in tokens for t in row))
M_fu = np.zeros((len(df), len(token_names)))
for r in range(len(df)):
    for t in tokens[r]:
        M_fu[r, token_names.index(t)] = 1
blocks['Follow-up tokens'] = (M_fu, token_names)
print("Symptom columns:", M_sym.shape[1], "| Follow-up token columns:", M_fu.shape[1])
print("Average symptoms per vehicle: %.2f | average follow-up tokens: %.2f" % (M_sym.sum(axis=1).mean(), M_fu.sum(axis=1).mean()))
```

```python
# 3) New numeric columns
df['Vehicle_Age'] = 2026 - df['Year']
df['Mileage_per_Year'] = df['Mileage_KM'] / df['Vehicle_Age']
print(df[['Year', 'Vehicle_Age', 'Mileage_KM', 'Mileage_per_Year']].describe().round(1).T)

# 4) TF-IDF of the complaint text (learned from the TRAIN rows only)
docs = [re.findall('[a-z]+', t.lower()) for t in df['Customer_Complaint_Text']]     # list of words of every complaint
doc_freq = Counter()
for i in train_idx:
    doc_freq.update(set(docs[i]))                       # in how many train complaints does each word appear?
vocab = sorted(w for w, c in doc_freq.items() if c >= 30)     # keep words that appear in at least 30 complaints
idf = np.array([np.log(len(train_idx) / doc_freq[w]) + 1 for w in vocab])

M_tfidf = np.zeros((len(df), len(vocab)))
for r in range(len(df)):
    counts = Counter(docs[r])
    for j, w in enumerate(vocab):
        if w in counts:
            M_tfidf[r, j] = counts[w] / len(docs[r]) * idf[j]            # tf x idf
M_tfidf = M_tfidf / np.linalg.norm(M_tfidf, axis=1, keepdims=True)       # L2 normalisation
blocks['TF-IDF of complaint'] = (M_tfidf, vocab)
print("TF-IDF words:", len(vocab))

# Note: Proof. The library TfidfVectorizer (same settings) must give the same numbers.
from sklearn.feature_extraction.text import TfidfVectorizer
tv = TfidfVectorizer(token_pattern='[a-z]+', vocabulary=vocab, smooth_idf=False, norm='l2')
tv.fit(df['Customer_Complaint_Text'].iloc[train_idx])
lib = tv.transform(df['Customer_Complaint_Text']).toarray()
print("Largest difference from the library:", np.abs(lib - M_tfidf).max())

# Does the complaint hold more than the symptoms? Check if it contains the text of all the symptoms.
inside = []
for r in range(len(df)):
    text = df.loc[r, 'Customer_Complaint_Text'].lower()
    inside.append(all(s.lower() in text for s in df.loc[r, symptom_cols] if s != 'No_Symptom'))
print("Complaints that contain all the symptom texts of the record: %.1f%%" % (100 * np.mean(inside)))
```

```python
# A table where every column is a category (numeric columns are cut into 5 bins using the train rows)
cat = pd.DataFrame(index=df.index)
for c in ['Vehicle_Type', 'Make', 'Model', 'Fuel', 'Transmission', 'Color', 'Registration_State', 'City',
          'Service_Center_Type', 'Warranty_Status', 'Insurance_Status', 'Service_History',
          'Symptom_1', 'Symptom_2', 'Symptom_3', 'Followup_Question_1', 'Followup_Answer_1',
          'Followup_Question_2', 'Followup_Answer_2']:
    cat[c] = df[c]
for c in ['Year', 'Mileage_KM', 'Engine_Displacement_CC', 'Fuel_Efficiency_kmpl', 'Vehicle_Age', 'Mileage_per_Year']:
    edges = np.unique(np.quantile(df.iloc[train_idx][c], [0.2, 0.4, 0.6, 0.8]))
    cat[c] = np.digitize(df[c], edges)
cat['Owner_Number'] = df['Owner_Number']
cat['Customer_Rating'] = df['Customer_Rating']
cat['Inspection_Month'] = df['Inspection_Date'].dt.month
cat['Followup_Pair_1'] = df['Followup_Question_1'] + ' | ' + df['Followup_Answer_1']
cat['Followup_Pair_2'] = df['Followup_Question_2'] + ' | ' + df['Followup_Answer_2']
cat['Symptom_Count'] = M_sym.sum(axis=1)

def mutual_information(x, y):
    table = pd.crosstab(x, y).to_numpy().astype(float)
    p = table / table.sum()                               # joint probability
    px = p.sum(axis=1, keepdims=True)
    py = p.sum(axis=0, keepdims=True)
    nonzero = p > 0
    return (p[nonzero] * np.log2(p[nonzero] / (px @ py)[nonzero])).sum()

train_cat = cat.iloc[train_idx]
y_tr_names = y_all[train_idx]
shuffler = np.random.default_rng(0)
rows = []
for c in cat.columns:
    real = mutual_information(train_cat[c].to_numpy(), y_tr_names)
    shuffled = np.mean([mutual_information(train_cat[c].to_numpy(), shuffler.permutation(y_tr_names)) for i in range(10)])
    rows.append({'column': c, 'levels': train_cat[c].nunique(), 'MI (bits)': round(real, 3),
                 'MI shuffled': round(shuffled, 3), 'real information': round(real - shuffled, 3)})
mi_table = pd.DataFrame(rows).sort_values('real information', ascending=False).reset_index(drop=True)
print(mi_table.to_string())
print("\nTotal information needed, H(Diagnosis) = %.3f bits" % mutual_information(y_tr_names, y_tr_names))

plt.figure(figsize=(8, 7))
top = mi_table.head(20).iloc[::-1]
plt.barh(top['column'], top['real information'])
plt.xlabel('Real information about Diagnosis (bits)')
plt.title('Filter method: top 20 columns')
plt.tight_layout()
plt.show()
```

```python
# Feature groups. Categorical columns -> one-hot (0/1 columns), numeric columns -> 5 bins -> one-hot.
def one_hot(series):
    table = pd.get_dummies(series.astype(str))
    return table.to_numpy(dtype=float), list(table.columns)

for name, cols in [('Vehicle_Type', ['Vehicle_Type']), ('Make', ['Make']), ('Model', ['Model']), ('Fuel', ['Fuel']),
                   ('Transmission', ['Transmission']), ('Year', ['Year']), ('Mileage_KM', ['Mileage_KM']),
                   ('Engine_CC', ['Engine_Displacement_CC']), ('Fuel_Efficiency', ['Fuel_Efficiency_kmpl']),
                   ('Vehicle_Age', ['Vehicle_Age']), ('Mileage_per_Year', ['Mileage_per_Year']),
                   ('Color', ['Color']), ('Owner_Number', ['Owner_Number']), ('State', ['Registration_State']),
                   ('City', ['City']), ('Service_Center', ['Service_Center_Type']), ('Warranty', ['Warranty_Status']),
                   ('Insurance', ['Insurance_Status']), ('Service_History', ['Service_History'])]:
    blocks[name] = one_hot(cat[cols[0]])

def bernoulli_nb(X_train, y_train, X_valid):
    X_train = (X_train > 0).astype(float)                  # presence of the column (TF-IDF values become 0/1 here)
    X_valid = (X_valid > 0).astype(float)
    onehot_y = (y_train[:, None] == np.arange(35)[None, :]).astype(float)     # rows x 35
    n_class = onehot_y.sum(axis=0)
    theta = (onehot_y.T @ X_train + 1) / (n_class[:, None] + 2)               # 35 x columns
    score = X_valid @ np.log(theta).T + (1 - X_valid) @ np.log(1 - theta).T + np.log(n_class / len(y_train))
    return score.argmax(axis=1)

# 5-fold cross-validation inside the train rows: the train rows are cut into 5 parts,
# each part is used once as the validation part while the other 4 parts are used to fit.
rng2 = np.random.default_rng(1)
folds = np.array_split(rng2.permutation(train_idx), 5)

def valid_accuracy(group_names):
    X = np.hstack([blocks[g][0] for g in group_names])
    accuracies = []
    for k in range(5):
        valid_rows = folds[k]
        fit_rows = np.concatenate([folds[j] for j in range(5) if j != k])
        pred = bernoulli_nb(X[fit_rows], y[fit_rows], X[valid_rows])
        accuracies.append((pred == y[valid_rows]).mean())
    return np.mean(accuracies)

# Each group alone (average accuracy of the 5 folds)
single = pd.Series({g: valid_accuracy([g]) for g in blocks}).sort_values(ascending=False)
print("Validation accuracy of each group alone (chance = 1/35 = 0.029, always the biggest class = 0.045):")
print(single.round(3).to_string())
```

```python
# Forward selection
chosen = []
best_so_far = 0.0
history = []
while True:
    left = [g for g in blocks if g not in chosen]
    scores = {g: valid_accuracy(chosen + [g]) for g in left}
    best_group = max(scores, key=scores.get)
    gain = scores[best_group] - best_so_far
    history.append({'round': len(chosen) + 1, 'best group to add': best_group,
                    'accuracy after adding': round(scores[best_group], 3), 'gain': round(gain, 3)})
    if gain < 0.01:                            # less than 1 percentage point -> stop
        break
    chosen.append(best_group)
    best_so_far = scores[best_group]
print(pd.DataFrame(history).to_string(index=False))
print("\nSelected groups:", chosen)
```

```python
final_groups = chosen
X_final = np.hstack([blocks[g][0] for g in final_groups])
feature_names = sum([blocks[g][1] for g in final_groups], [])
X_train, X_test = X_final[train_idx], X_final[test_idx]
print("Final groups:", final_groups)
print("Final matrix: train", X_train.shape, "| test", X_test.shape)

# Scaled vehicle numbers (train mean and std only)
num_cols = ['Year', 'Mileage_KM', 'Engine_Displacement_CC', 'Fuel_Efficiency_kmpl']
train_mean = df.iloc[train_idx][num_cols].mean()
train_std = df.iloc[train_idx][num_cols].std(ddof=0)
Znum = ((df[num_cols] - train_mean) / train_std).to_numpy()
print("Mean of scaled train columns:", Znum[train_idx].mean(axis=0).round(4))
print("Std of scaled train columns :", Znum[train_idx].std(axis=0).round(4))

# Save for later use
np.savez_compressed('model_data.npz', X_train=X_train, X_test=X_test, y_train=y_train, y_test=y_test)
print("Saved model_data.npz")
```

```python
def pca(X):
    centred = X - X.mean(axis=0)
    cov = centred.T @ centred / (len(X) - 1)
    values, vectors = np.linalg.eigh(cov)              # eigh = eigenvalues of a symmetric matrix
    order = np.argsort(-values)                        # biggest first
    return values[order], vectors[:, order]

# (a) vehicle numbers (4 columns, scaled)
values4, vectors4 = pca(Znum[train_idx])
print("Explained share of PC1..PC4:", (values4 / values4.sum()).round(3))
print("PC1 loadings (", num_cols, "):", vectors4[:, 0].round(2))

# (b) final feature matrix
values_f, vectors_f = pca(X_train)
ratio = values_f / values_f.sum()
cumulative = np.cumsum(ratio)
for target in [0.5, 0.8, 0.9]:
    print("Components needed for %d%% of the spread: %d (of %d columns)" % (target * 100, np.argmax(cumulative >= target) + 1, X_train.shape[1]))

plt.figure(figsize=(6, 3.5))
plt.plot(range(1, 41), cumulative[:40], marker='o', markersize=3)
plt.xlabel('Number of components'); plt.ylabel('Cumulative explained share'); plt.title('PCA of the final feature matrix')
plt.show()

# Note: Proof. The library PCA must give the same explained shares.
from sklearn.decomposition import PCA
print("Largest difference from the library (4 numbers):", np.abs(PCA().fit(Znum[train_idx]).explained_variance_ratio_ - values4 / values4.sum()).max())
print("Largest difference from the library (final matrix, first 10):",
      np.abs(PCA().fit(X_train).explained_variance_ratio_[:10] - ratio[:10]).max())
```

# OPTIMISATION BASICS

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv('vehicle_diagnostic_clean.csv')

# Scaled Year (x) and Mileage_KM (y) of the train rows only (train_idx was made in the feature engineering step)
x = ((df['Year'] - df['Year'].iloc[train_idx].mean()) / df['Year'].iloc[train_idx].std(ddof=0)).to_numpy()[train_idx]
y_m = ((df['Mileage_KM'] - df['Mileage_KM'].iloc[train_idx].mean()) / df['Mileage_KM'].iloc[train_idx].std(ddof=0)).to_numpy()[train_idx]

def mse(w):
    return np.mean((y_m - w * x) ** 2)

def slope(w):
    return -2 * np.mean(x * (y_m - w * x))         # first derivative of MSE with respect to w

# Where is the slope 0? Solve slope(w) = 0:  -2 mean(x y) + 2 w mean(x^2) = 0   ->   w = mean(x y) / mean(x^2)
w_best = np.mean(x * y_m) / np.mean(x ** 2)
print("Best w from slope = 0 :", round(w_best, 4))
print("Slope at best w       :", round(slope(w_best), 10))
print("Second derivative     :", round(2 * np.mean(x ** 2), 4), "(positive -> it is a minimum)")

# Note: Proof. A brute force search over many w values must find the same minimum.
grid = np.linspace(-2, 1, 3001)
print("Best w from brute force:", round(grid[np.argmin([mse(w) for w in grid])], 4))

plt.figure(figsize=(6, 3.5))
plt.plot(grid, [mse(w) for w in grid])
plt.scatter([w_best], [mse(w_best)], color='red', zorder=3, label='minimum')
plt.xlabel('w'); plt.ylabel('MSE(w)'); plt.title('Error against the parameter w'); plt.legend()
plt.show()
```

```python
def gradient_descent(lr, steps=30):
    w = -1.5                                       # a starting point far from the minimum
    history = [mse(w)]
    for step in range(steps):
        w = w - lr * slope(w)                      # one downhill step
        history.append(mse(w))
    return w, history

plt.figure(figsize=(6, 3.5))
for lr in [0.05, 0.5, 1.1]:
    w_final, history = gradient_descent(lr)
    plt.plot(history, label='lr = %.2f' % lr)
    print("lr = %.2f -> final w = %.4f, final MSE = %.4f" % (lr, w_final, history[-1]))
plt.yscale('log')
plt.xlabel('Step'); plt.ylabel('MSE'); plt.title('Gradient descent with 3 learning rates'); plt.legend()
plt.show()
print("Best w from Step 1:", round(w_best, 4))
```

```python
# Real units: Year -> Mileage_KM (train rows)
year = df['Year'].to_numpy(dtype=float)[train_idx]
km = df['Mileage_KM'].to_numpy(dtype=float)[train_idx]

w = np.sum((year - year.mean()) * (km - km.mean())) / np.sum((year - year.mean()) ** 2)
b = km.mean() - w * year.mean()
print("Least squares: Mileage = %.1f x Year + %.1f" % (w, b))
print("Meaning: each newer year lowers the mileage by about %.0f km" % abs(w))

# Normal equation with a column of ones for b
A = np.column_stack([year, np.ones(len(year))])
print("Normal equation:", np.linalg.solve(A.T @ A, A.T @ km).round(1))

# Note: Proof. Library results must be the same.
print("np.polyfit     :", np.polyfit(year, km, 1).round(1))
# Gradient descent on the same problem (scaled data) must reach the same line
w_gd, _ = gradient_descent(0.5, steps=200)
print("Gradient descent w (scaled): %.4f | formula w (scaled): %.4f" % (w_gd, w_best))
```

```python
def f(p):
    return (p[0] - 3) ** 2 + (p[1] - 2) ** 2

# Lagrange solution (calculated by hand above)
print("Lagrange solution: x = 1.5, y = 0.5, f =", f([1.5, 0.5]))

# Projected gradient descent
p = np.array([0.0, 2.0])                            # start on the line x + y = 2
for step in range(200):
    gradient = np.array([2 * (p[0] - 3), 2 * (p[1] - 2)])
    p = p - 0.1 * gradient                          # normal downhill step
    p = p - (p.sum() - 2) / 2                       # move back to the line x + y = 2
print("Projected gradient descent:", p.round(4), "f =", round(f(p), 4))

# Note: Proof. The library must find the same point.
from scipy.optimize import minimize
result = minimize(f, x0=[0, 0], constraints={'type': 'eq', 'fun': lambda p: p[0] + p[1] - 2})
print("Library (scipy minimize):", result.x.round(4))

# Without the rule the best point would be (3, 2) with f = 0. The rule makes the best value worse:
print("Best f without the rule: 0 | best f with the rule:", f([1.5, 0.5]))
```

```python
def toy(a, b):
    # a bumpy bowl: its lowest point is near (1.3, -0.7), with small bumps that can trap a search
    return (a - 1.3) ** 2 + (b + 0.7) ** 2 + 0.3 * np.sin(5 * a) * np.sin(5 * b)

# Grid search: 10 x 10 = 100 points
a_values = np.linspace(-3, 3, 10)
b_values = np.linspace(-3, 3, 10)
grid_best = min((toy(a, b), a, b) for a in a_values for b in b_values)

# Random search: also 100 points
rng = np.random.default_rng(0)
random_points = rng.uniform(-3, 3, size=(100, 2))
random_best = min((toy(a, b), a, b) for a, b in random_points)

print("Grid search  : value %.3f at a = %.2f, b = %.2f" % grid_best)
print("Random search: value %.3f at a = %.2f, b = %.2f" % random_best)

# Note: Proof. A very fine grid (1000 x 1000 points) gives the true lowest value.
fine = np.linspace(-3, 3, 1000)
A_, B_ = np.meshgrid(fine, fine)
print("Very fine search: value %.3f" % toy(A_, B_).min())
```

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

# Helper functions used by all the models of this part
def accuracy(y_true, scores):
    return (scores.argmax(axis=1) == y_true).mean()               # the class with the highest score is the answer

def top_k_accuracy(y_true, scores, k=3):
    top = np.argsort(-scores, axis=1)[:, :k]                      # the k best classes of every row
    return (top == y_true[:, None]).any(axis=1).mean()            # is the true class among them?

def macro_f1(y_true, scores):
    y_pred = scores.argmax(axis=1)
    f1 = []
    for c in range(35):
        tp = np.sum((y_pred == c) & (y_true == c))
        fp = np.sum((y_pred == c) & (y_true != c))
        fn = np.sum((y_pred != c) & (y_true == c))
        precision = tp / (tp + fp) if tp + fp > 0 else 0
        recall = tp / (tp + fn) if tp + fn > 0 else 0
        f1.append(2 * precision * recall / (precision + recall) if precision + recall > 0 else 0)
    return np.mean(f1)                                            # every class counts the same

def r_squared(y_true, y_pred):
    return 1 - np.sum((y_true - y_pred) ** 2) / np.sum((y_true - y_true.mean()) ** 2)
```

```python
year = df['Year'].to_numpy(dtype=float)
km = df['Mileage_KM'].to_numpy(dtype=float)

# Fit on train rows
x_tr, y_tr = year[train_idx], km[train_idx]
slope_w = np.sum((x_tr - x_tr.mean()) * (y_tr - y_tr.mean())) / np.sum((x_tr - x_tr.mean()) ** 2)
intercept = y_tr.mean() - slope_w * x_tr.mean()

# Test on test rows
pred = slope_w * year[test_idx] + intercept
print("Line: Mileage_KM = %.1f x Year + %.1f" % (slope_w, intercept))
print("R-squared  train: %.4f | test: %.4f" % (r_squared(y_tr, slope_w * x_tr + intercept), r_squared(km[test_idx], pred)))
print("RMSE test: %.0f km | MAE test: %.0f km | std of Mileage_KM: %.0f km" % (
    np.sqrt(np.mean((km[test_idx] - pred) ** 2)), np.mean(np.abs(km[test_idx] - pred)), km.std()))

# Is the slope different from 0? (t test)
residual_ss = np.sum((y_tr - (slope_w * x_tr + intercept)) ** 2)
se_slope = np.sqrt(residual_ss / (len(x_tr) - 2) / np.sum((x_tr - x_tr.mean()) ** 2))
from scipy import stats
t_value = slope_w / se_slope
print("t = %.1f, p-value = %.3g" % (t_value, 2 * stats.t.sf(abs(t_value), len(x_tr) - 2)))

# Note: Proof. The library must give the same line.
from sklearn.linear_model import LinearRegression
lib = LinearRegression().fit(x_tr.reshape(-1, 1), y_tr)
print("Library: slope %.1f, intercept %.1f, test R-squared %.4f" % (lib.coef_[0], lib.intercept_, lib.score(year[test_idx].reshape(-1, 1), km[test_idx])))

# Graph: points and the line (test rows)
plt.figure(figsize=(6, 4))
plt.scatter(year[test_idx], km[test_idx], s=5, alpha=0.4)
xs = np.array([year.min(), year.max()])
plt.plot(xs, slope_w * xs + intercept, color='red')
plt.xlabel('Year'); plt.ylabel('Mileage_KM'); plt.title('Simple linear regression (test rows)')
plt.show()
```

```python
hours = df['Estimated_Repair_Time_Hours'].to_numpy()
target = np.log1p(hours)                                   # log(1 + hours)

lam = 1.0
x_mean = X_train.mean(axis=0)
Xc = X_train - x_mean                                       # centred inputs
t_mean = target[train_idx].mean()
w_ridge = np.linalg.solve(Xc.T @ Xc + lam * np.eye(X_train.shape[1]), Xc.T @ (target[train_idx] - t_mean))
b_ridge = t_mean - x_mean @ w_ridge

pred_log = X_test @ w_ridge + b_ridge
pred_hours = np.expm1(pred_log)
true_hours = hours[test_idx]
baseline = np.median(hours[train_idx])                      # always say the median repair time

print("R-squared on the log scale (test): %.3f" % r_squared(target[test_idx], pred_log))
print("MAE in hours (test): model %.3f | always-median baseline %.3f" % (np.mean(np.abs(true_hours - pred_hours)), np.mean(np.abs(true_hours - baseline))))
print("Average error in hours by severity of the true diagnosis:")
sev_test = df['Severity_Level'].iloc[test_idx].to_numpy()
for s in ['Low', 'Medium', 'High', 'Critical']:
    m = sev_test == s
    print("  %-8s true mean %.2f h | predicted mean %.2f h | MAE %.2f h" % (s, true_hours[m].mean(), pred_hours[m].mean(), np.abs(true_hours[m] - pred_hours[m]).mean()))

# Note: Proof. The library Ridge (same lambda) must give the same weights.
from sklearn.linear_model import Ridge
lib = Ridge(alpha=lam).fit(X_train, target[train_idx])
print("Largest difference of the weights from the library:", np.abs(lib.coef_ - w_ridge).max())

plt.figure(figsize=(5, 4))
plt.scatter(true_hours, pred_hours, s=5, alpha=0.4)
plt.plot([0, 30], [0, 30], color='red')
plt.xlabel('True repair hours'); plt.ylabel('Predicted repair hours'); plt.title('Repair time (test rows)')
plt.show()
```

# LOGISTIC REGRESSION

```python
def softmax(S):
    S = S - S.max(axis=1, keepdims=True)            # subtract the largest score: same result, no overflow
    E = np.exp(S)
    return E / E.sum(axis=1, keepdims=True)

def train_logistic(X, y, l2=1e-3, lr=0.5, steps=500):
    n, F = X.shape
    Y = (y[:, None] == np.arange(35)[None, :]).astype(float)      # true class as 0/1 columns
    W = np.zeros((F, 35))
    b = np.zeros(35)
    losses = []
    for step in range(steps):
        P = softmax(X @ W + b)
        losses.append(-np.mean(np.log(P[np.arange(n), y] + 1e-12)) + l2 / 2 * np.sum(W ** 2))
        W = W - lr * (X.T @ (P - Y) / n + l2 * W)                 # gradient descent step
        b = b - lr * (P - Y).mean(axis=0)
    return W, b, losses

# Note: Proof 1. Check the gradient formula with a tiny change of one weight (numerical derivative).
Xs, ys = X_train[:500], y_train[:500]
W0 = np.random.default_rng(0).normal(0, 0.1, size=(X_train.shape[1], 35))
def loss_of(W):
    P = softmax(Xs @ W)
    return -np.mean(np.log(P[np.arange(500), ys] + 1e-12)) + 1e-3 / 2 * np.sum(W ** 2)
P0 = softmax(Xs @ W0)
Y0 = (ys[:, None] == np.arange(35)[None, :]).astype(float)
gradient = Xs.T @ (P0 - Y0) / 500 + 1e-3 * W0
i, j, h = 3, 7, 1e-5
W_plus = W0.copy(); W_plus[i, j] += h
W_minus = W0.copy(); W_minus[i, j] -= h
print("Gradient from formula : %.8f" % gradient[i, j])
print("Gradient from numbers : %.8f" % ((loss_of(W_plus) - loss_of(W_minus)) / (2 * h)))
```

```python
# Choose the L2 strength by 5-fold cross-validation on the train rows (grid search, Part H Step 5)
train_pos = {row: p for p, row in enumerate(train_idx)}         # position of a row inside X_train
fold_pos = [np.array([train_pos[r] for r in f]) for f in folds]

def cv_accuracy_logistic(l2):
    scores = []
    for k in range(5):
        valid = fold_pos[k]
        fit = np.concatenate([fold_pos[j] for j in range(5) if j != k])
        W, b, _ = train_logistic(X_train[fit], y_train[fit], l2=l2)
        scores.append(accuracy(y_train[valid], X_train[valid] @ W + b))
    return np.mean(scores)

cv_result = {}
for l2 in [1e-4, 1e-3, 1e-2]:
    cv_result[l2] = cv_accuracy_logistic(l2)
    print("L2 = %g: 5-fold accuracy = %.4f" % (l2, cv_result[l2]))
best_l2 = max(cv_result, key=cv_result.get)          # the L2 strength with the best accuracy
print("Chosen L2:", best_l2)
```

```python
W_log, b_log, losses = train_logistic(X_train, y_train, l2=best_l2)
scores_log = softmax(X_test @ W_log + b_log)              # probabilities of the test rows

print("Loss at the start %.3f -> at the end %.3f (must go down)" % (losses[0], losses[-1]))
print("Loss never went up:", all(losses[i + 1] <= losses[i] + 1e-12 for i in range(len(losses) - 1)))
print("Test accuracy: %.4f | top-3 accuracy: %.4f | macro F1: %.4f" % (
    accuracy(y_test, scores_log), top_k_accuracy(y_test, scores_log), macro_f1(y_test, scores_log)))

# Note: Proof 2. The library logistic regression (similar penalty) should give a similar accuracy.
from sklearn.linear_model import LogisticRegression
lib = LogisticRegression(C=1 / (len(y_train) * best_l2), max_iter=2000).fit(X_train, y_train)
print("Library test accuracy: %.4f" % lib.score(X_test, y_test))

plt.figure(figsize=(6, 3.5))
plt.plot(losses)
plt.xlabel('Step'); plt.ylabel('Loss'); plt.title('Logistic regression: loss during training')
plt.show()
```

# k-NEAREST NEIGHBOURS (kNN)

```python
def squared_distances(A, B):
    # squared distance of every row of A to every row of B
    return (A ** 2).sum(axis=1)[:, None] + (B ** 2).sum(axis=1)[None, :] - 2 * A @ B.T

def knn_scores(X_fit, y_fit, X_new, k):
    D = squared_distances(X_new, X_fit)
    nearest = np.argsort(D, axis=1)[:, :k]                  # positions of the k nearest train rows
    labels = y_fit[nearest]                                 # their diagnoses
    votes = np.zeros((len(X_new), 35))
    for c in range(35):
        votes[:, c] = (labels == c).sum(axis=1)             # how many neighbours vote for class c
    return votes / k

# Grid search for k
rows = []
for k in [1, 3, 5, 10, 15, 25, 40, 60]:
    acc = []
    for f in range(5):
        valid = fold_pos[f]
        fit = np.concatenate([fold_pos[j] for j in range(5) if j != f])
        acc.append(accuracy(y_train[valid], knn_scores(X_train[fit], y_train[fit], X_train[valid], k)))
    rows.append({'k': k, '5-fold accuracy': round(np.mean(acc), 4)})
knn_table = pd.DataFrame(rows)
print(knn_table.to_string(index=False))
best_k = int(knn_table.loc[knn_table['5-fold accuracy'].idxmax(), 'k'])
print("Chosen k:", best_k)
```

```python
scores_knn = knn_scores(X_train, y_train, X_test, best_k)
print("Test accuracy: %.4f | top-3 accuracy: %.4f | macro F1: %.4f" % (
    accuracy(y_test, scores_knn), top_k_accuracy(y_test, scores_knn), macro_f1(y_test, scores_knn)))

# Note: Proof. The library kNN with the same k must give about the same accuracy
# (small differences come from how ties between equally near vehicles are broken).
from sklearn.neighbors import KNeighborsClassifier
lib = KNeighborsClassifier(n_neighbors=best_k).fit(X_train, y_train)
print("Library test accuracy: %.4f" % lib.score(X_test, y_test))

plt.figure(figsize=(5, 3.5))
plt.plot(knn_table['k'], knn_table['5-fold accuracy'], marker='o')
plt.xlabel('k'); plt.ylabel('5-fold accuracy'); plt.title('kNN: choosing k')
plt.show()
```

# DECISION TREE AND RANDOM FOREST

```python
def gini(counts):
    return 1 - np.sum((counts / counts.sum()) ** 2)

def build_tree(X, Yoh, rows, depth, max_depth, rng=None, n_try=None):
    # X = 0/1 matrix, Yoh = true class as 0/1 columns, rows = the rows that reached this node
    counts = Yoh[rows].sum(axis=0)                               # how many rows of each diagnosis are here
    node = {'counts': counts}                                    # a leaf keeps only the counts
    n = len(rows)
    if depth >= max_depth or n < 2 or (counts > 0).sum() == 1:   # too deep, too small or already pure
        return node

    Xr = X[rows]
    n_yes = Xr.sum(axis=0)                                       # rows with the column = 1
    n_no = n - n_yes
    counts_yes = Yoh[rows].T @ Xr                                # class counts for "yes" (35 x columns)
    counts_no = counts[:, None] - counts_yes                     # class counts for "no"
    gini_yes = 1 - ((counts_yes / np.maximum(n_yes, 1)) ** 2).sum(axis=0)
    gini_no = 1 - ((counts_no / np.maximum(n_no, 1)) ** 2).sum(axis=0)
    weighted = (n_yes * gini_yes + n_no * gini_no) / n           # weighted Gini of every possible question
    weighted[(n_yes == 0) | (n_no == 0)] = np.inf                # a question that does not split is useless

    if n_try is not None:                                        # random forest: only a random set of columns is allowed
        allowed = rng.choice(X.shape[1], size=n_try, replace=False)
        blocked = np.ones(X.shape[1], dtype=bool)
        blocked[allowed] = False
        weighted[blocked] = np.inf

    best = int(np.argmin(weighted))
    if weighted[best] >= gini(counts) - 1e-12:                   # no question makes the groups purer -> leaf
        return node
    yes = Xr[:, best] == 1
    node['column'] = best
    node['yes'] = build_tree(X, Yoh, rows[yes], depth + 1, max_depth, rng, n_try)
    node['no'] = build_tree(X, Yoh, rows[~yes], depth + 1, max_depth, rng, n_try)
    return node

def tree_scores(node, X, rows=None, out=None):
    # send the rows down the tree; a leaf gives the class shares of its train rows
    if out is None:
        out = np.zeros((len(X), 35))
        rows = np.arange(len(X))
    if 'column' not in node:
        out[rows] = node['counts'] / node['counts'].sum()
        return out
    yes = X[rows, node['column']] == 1
    tree_scores(node['yes'], X, rows[yes], out)
    tree_scores(node['no'], X, rows[~yes], out)
    return out

def count_leaves(node):
    return 1 if 'column' not in node else count_leaves(node['yes']) + count_leaves(node['no'])

Y_train = (y_train[:, None] == np.arange(35)[None, :]).astype(float)

# Note: Proof. Gini of a pure group is 0, of two equal classes 0.5, of 35 equal classes 1 - 1/35.
print("Gini pure: %.3f | two equal classes: %.3f | 35 equal classes: %.4f (1 - 1/35 = %.4f)" % (
    gini(np.array([10, 0])), gini(np.array([5, 5])), gini(np.ones(35)), 1 - 1 / 35))
```

```python
# Grid search for the depth, 5-fold cross-validation on the train rows
rows = []
for depth in [5, 10, 20, 40, 70, 100]:
    acc = []
    for f in range(5):
        valid = fold_pos[f]
        fit = np.concatenate([fold_pos[j] for j in range(5) if j != f])
        tree = build_tree(X_train, Y_train, fit, 0, depth)
        acc.append(accuracy(y_train[valid], tree_scores(tree, X_train[valid])))
    rows.append({'max depth': depth, '5-fold accuracy': round(np.mean(acc), 4)})
tree_table = pd.DataFrame(rows)
print(tree_table.to_string(index=False))
best_depth = int(tree_table.loc[tree_table['5-fold accuracy'].idxmax(), 'max depth'])
print("Chosen max depth:", best_depth)
```

```python
tree = build_tree(X_train, Y_train, np.arange(len(X_train)), 0, best_depth)
scores_tree = tree_scores(tree, X_test)
print("Leaves:", count_leaves(tree))
print("Train accuracy: %.4f | test accuracy: %.4f | top-3: %.4f | macro F1: %.4f" % (
    accuracy(y_train, tree_scores(tree, X_train)), accuracy(y_test, scores_tree),
    top_k_accuracy(y_test, scores_tree), macro_f1(y_test, scores_tree)))

# Note: Proof. The library tree (same Gini rule and depth) must give a similar accuracy.
# (not exactly the same, because when two questions are equally good the library may pick the other one).
from sklearn.tree import DecisionTreeClassifier
lib = DecisionTreeClassifier(criterion='gini', max_depth=best_depth, random_state=0).fit(X_train, y_train)
print("Library test accuracy: %.4f" % lib.score(X_test, y_test))

# The first questions of the tree, in words
def show_tree(node, depth=0, max_depth=3):
    if 'column' not in node or depth >= max_depth:
        top = int(node['counts'].argmax())
        print("  " * depth + "-> diagnosis:", classes[top], "(%d of %d rows)" % (node['counts'][top], node['counts'].sum()))
        return
    print("  " * depth + "Is '" + feature_names[node['column']] + "' present?")
    print("  " * depth + " YES:")
    show_tree(node['yes'], depth + 1, max_depth)
    print("  " * depth + " NO:")
    show_tree(node['no'], depth + 1, max_depth)
show_tree(tree, 0, 2)

plt.figure(figsize=(5, 3.5))
plt.plot(tree_table['max depth'], tree_table['5-fold accuracy'], marker='o')
plt.xlabel('Max depth'); plt.ylabel('5-fold accuracy'); plt.title('Decision tree: choosing the depth')
plt.show()
```

```python
# Random forest: 50 trees, each on a bootstrap sample, each question may look at sqrt(columns) random columns
n_trees = 50
n_try = int(np.sqrt(X_train.shape[1]))
rng = np.random.default_rng(0)
forest = []
for t in range(n_trees):
    sample = rng.integers(0, len(X_train), size=len(X_train))        # draw rows WITH replacement
    forest.append(build_tree(X_train, Y_train, sample, 0, 100, rng, n_try))

def forest_scores(forest, X):
    return np.mean([tree_scores(t, X) for t in forest], axis=0)      # average of the class shares of all trees

scores_rf = forest_scores(forest, X_test)
print("Random forest: %d trees, %d columns tried per question" % (n_trees, n_try))
print("Test accuracy: %.4f | top-3: %.4f | macro F1: %.4f" % (
    accuracy(y_test, scores_rf), top_k_accuracy(y_test, scores_rf), macro_f1(y_test, scores_rf)))

# Note: Proof. The library forest with the same settings must give a similar accuracy.
from sklearn.ensemble import RandomForestClassifier
lib = RandomForestClassifier(n_estimators=n_trees, max_features='sqrt', random_state=0).fit(X_train, y_train)
print("Library test accuracy: %.4f" % lib.score(X_test, y_test))

# How does the accuracy grow with the number of trees? (shown for understanding, the number of trees was fixed before)
counts = [1, 5, 10, 20, 35, 50]
curve = [accuracy(y_test, forest_scores(forest[:c], X_test)) for c in counts]
print(dict(zip(counts, np.round(curve, 4))))
plt.figure(figsize=(5, 3.5))
plt.plot(counts, curve, marker='o')
plt.xlabel('Number of trees'); plt.ylabel('Test accuracy'); plt.title('Random forest')
plt.show()
```

# SUPPORT VECTOR MACHINE (SVM)

```python
# 1) Find the two diagnoses that logistic regression confuses most (on the train rows)
train_pred = (X_train @ W_log + b_log).argmax(axis=1)
confusion_train = np.zeros((35, 35))
for t, p in zip(y_train, train_pred):
    confusion_train[t, p] += 1
pair_errors = confusion_train + confusion_train.T                 # errors in both directions
np.fill_diagonal(pair_errors, 0)
a, b_ = np.unravel_index(pair_errors.argmax(), pair_errors.shape)
print("Hardest pair:", classes[a], "VS", classes[b_], "| train errors between them:", int(pair_errors[a, b_]))
print("System categories:", df.loc[df['Diagnosis'] == classes[a], 'System_Category'].iloc[0], "and", df.loc[df['Diagnosis'] == classes[b_], 'System_Category'].iloc[0])

# 2) Binary data: label +1 for the first diagnosis, -1 for the second
def binary_data(X, yy):
    keep = (yy == a) | (yy == b_)
    return X[keep], np.where(yy[keep] == a, 1.0, -1.0)
Xb_train, yb_train = binary_data(X_train, y_train)
Xb_test, yb_test = binary_data(X_test, y_test)
print("Train rows:", len(Xb_train), "| test rows:", len(Xb_test))
```

```python
def train_svm(X, label, lam=0.01, steps=1000):
    w = np.zeros(X.shape[1])
    b = 0.0
    for t in range(1, steps + 1):
        margin = label * (X @ w + b)
        wrong = margin < 1                                        # points that break the rule
        grad_w = lam * w - (label[wrong, None] * X[wrong]).sum(axis=0) / len(X)
        grad_b = -label[wrong].sum() / len(X)
        step = 1 / (lam * t)                                      # step size becomes smaller with time
        w = w - step * grad_w
        b = b - step * grad_b
    return w, b

lam = 0.01
w_svm, b_svm = train_svm(Xb_train, yb_train, lam)
pred_train = np.sign(Xb_train @ w_svm + b_svm)
pred_test = np.sign(Xb_test @ w_svm + b_svm)
print("Our SVM: train accuracy %.4f | test accuracy %.4f" % ((pred_train == yb_train).mean(), (pred_test == yb_test).mean()))
print("Margin width 2/||w|| = %.3f" % (2 / np.linalg.norm(w_svm)))

# For comparison: logistic regression on the same two diagnoses (the 35-class model, test rows of these two classes)
keep = (y_test == a) | (y_test == b_)
two_class = np.where((X_test[keep] @ W_log + b_log)[:, a] > (X_test[keep] @ W_log + b_log)[:, b_], 1.0, -1.0)
print("35-class logistic regression on these two diagnoses: test accuracy %.4f" % (two_class == yb_test).mean())

# Note: Proof. The library SVM (linear kernel, same penalty) must give a similar accuracy.
from sklearn.svm import SVC
lib = SVC(kernel='linear', C=1 / (lam * len(Xb_train))).fit(Xb_train, yb_train)
print("Library SVM: train accuracy %.4f | test accuracy %.4f | support vectors %d" % (
    lib.score(Xb_train, yb_train), lib.score(Xb_test, yb_test), lib.n_support_.sum()))
```

```python
def cross_validate(train_function):
    accs = []
    for f in range(5):
        valid = fold_pos[f]
        fit = np.concatenate([fold_pos[j] for j in range(5) if j != f])
        accs.append(accuracy(y_train[valid], train_function(fit, valid)))
    return np.mean(accs), np.std(accs)

def run_logistic(fit, valid):
    W, b, _ = train_logistic(X_train[fit], y_train[fit], l2=best_l2)
    return X_train[valid] @ W + b

def run_knn(fit, valid):
    return knn_scores(X_train[fit], y_train[fit], X_train[valid], best_k)

def run_tree(fit, valid):
    t = build_tree(X_train, Y_train, fit, 0, best_depth)
    return tree_scores(t, X_train[valid])

def run_forest(fit, valid):
    r = np.random.default_rng(0)
    trees = []
    for i in range(20):                                          # 20 trees per fold, to keep the time short
        sample = fit[r.integers(0, len(fit), size=len(fit))]
        trees.append(build_tree(X_train, Y_train, sample, 0, 100, r, n_try))
    return forest_scores(trees, X_train[valid])

cv_rows = []
for name, fn in [('Logistic Regression', run_logistic), ('kNN', run_knn), ('Decision Tree', run_tree), ('Random Forest (20 trees)', run_forest)]:
    mean, std = cross_validate(fn)
    cv_rows.append({'model': name, 'CV accuracy (mean)': round(mean, 4), 'CV std': round(std, 4)})
print(pd.DataFrame(cv_rows).to_string(index=False))
```
